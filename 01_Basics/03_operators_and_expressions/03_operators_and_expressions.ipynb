## 1. Core Definitions & Operator Taxonomy

### What is an Operator?
> **Definition**: An **Operator** is a predefined symbol that instructs the Python Virtual Machine (PVM) to perform a specific computation, comparison, or data manipulation on one or more operands (literals, variables, expressions) and return a resulting value.

### What is an Expression?
> **Definition**: An **Expression** is any valid combination of variables, constants, literals, and function calls joined by operators that the Python interpreter evaluates to produce a single value.

### The 7 Major Operator Categories in Python:
1. **Arithmetic Operators**: `+`, `-`, `*`, `/`, `//`, `%`, `**`
2. **Assignment & Augmented Operators**: `=`, `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, `**=`, Walrus `:=`
3. **Relational (Comparison) Operators**: `>`, `<`, `==`, `!=`, `>=`, `<=`
4. **Logical Operators**: `and`, `or`, `not`
5. **Bitwise Operators**: `&`, `|`, `^`, `~`, `<<`, `>>`
6. **Identity Operators**: `is`, `is not`
7. **Membership Operators**: `in`, `not in`

## 3. The 4 Variable Swapping Techniques (Interview Comparative Analysis)

Swapping the values of two variables is a classic computer science problem. In Python, there are **4 distinct approaches**:

1. **Approach 1: Using a Third Temporary Variable**: Universal across all languages (C, C++, Java). Safe and intuitive.
2. **Approach 2: Using Arithmetic Addition & Subtraction (`+`, `-`)**: Eliminates the extra variable, but can lead to integer overflow in low-level languages (though safe in Python's arbitrary-precision integers).
3. **Approach 3: Using Bitwise XOR (`^`)**: Highly performant low-level manipulation. Works because $x \oplus x = 0$ and $x \oplus 0 = x$.
4. **Approach 4: Pythonic Tuple Packing & Unpacking**: Idiomatic, optimized at the bytecode level using `ROT_TWO` / tuple assignment.

```python
# The 4 Swapping Techniques Implementation Drill

# Technique 1: Using a third temporary variable
x, y = 10, 20
temp = x
x = y
y = temp
print(f"Technique 1 (Temp Var):     x = {x}, y = {y}")

# Technique 2: Using Arithmetic (+ and -) (Without third variable)
x, y = 10, 20
x = x + y  # x = 30
y = x - y  # y = 10
x = x - y  # x = 20
print(f"Technique 2 (Arithmetic):   x = {x}, y = {y}")

# Technique 3: Using Bitwise XOR (^) (No third variable)
x, y = 10, 20
x = x ^ y
y = x ^ y
x = x ^ y
print(f"Technique 3 (Bitwise XOR):  x = {x}, y = {y}")

# Technique 4: Pythonic Multi-Assignment (Tuple Unpacking)
x, y = 10, 20
x, y = y, x
print(f"Technique 4 (Pythonic):     x = {x}, y = {y}")
```

## 4. Bitwise Operators Bit-by-Bit Deep Dive

> **Fundamental Principle**: Bitwise operators evaluate numbers **bit-by-bit** on their binary representations.
> **Why do bitwise operators only work on integers?**
> Integers possess exact binary certainty with fixed two's-complement representation. Floating-point numbers use IEEE 754 mantissa/exponent approximations without discrete bit positions.

| Operator | Name | Bit Rule | Mathematical Analogy |
|---|---|---|---|
| `&` | Bitwise AND | 1 if **both** bits are 1; else 0 | Logical intersection |
| `\|` | Bitwise OR | 1 if **at least one** bit is 1; else 0 | Logical union |
| `^` | Bitwise XOR | 1 if bits **differ**; 0 if bits match | Toggle / Parity check |
| `~` | Bitwise NOT | Inverts every bit ($0 \rightarrow 1, 1 \rightarrow 0$) | Formula: `~x = -(x + 1)` |
| `<<`| Left Shift | Shifts bits left, fills zeros on right | Multiplies by $2^k$: `x << k = x * (2**k)` |
| `>>`| Right Shift | Shifts bits right, discards lowest bits | Floor divides by $2^k$: `x >> k = x // (2**k)` |

```python
# Bitwise Operators Execution & Binary Bit Breakdown
a = 12   # Binary: 0000 1100
b = 10   # Binary: 0000 1010

print(f"a = {a:2d} -> Binary: {a:08b}")
print(f"b = {b:2d} -> Binary: {b:08b}")
print("-" * 35)

# 1. Bitwise AND (&)
and_res = a & b
print(f"a & b   = {and_res:2d} -> Binary: {and_res:08b}")

# 2. Bitwise OR (|)
or_res = a | b
print(f"a | b   = {or_res:2d} -> Binary: {or_res:08b}")

# 3. Bitwise XOR (^)
xor_res = a ^ b
print(f"a ^ b   = {xor_res:2d} -> Binary: {xor_res:08b}")

# 4. Bitwise NOT (~) formula: ~x = -(x + 1)
not_res = ~a
print(f"~a      = {not_res:2d} (Formula -(12 + 1) = {-(a + 1)})")

# 5. Bitwise Left Shift (<<)
shift_left = a << 2  # 12 * (2^2) = 48
print(f"a << 2  = {shift_left:2d} (12 * 2^2 = {a * (2**2)}) -> Binary: {shift_left:08b}")

# 6. Bitwise Right Shift (>>)
shift_right = a >> 2 # 12 // (2^2) = 3
print(f"a >> 2  = {shift_right:2d} (12 // 2^2 = {a // (2**2)}) -> Binary: {shift_right:08b}")
```

## 5. Ternary Operator (Conditional Expressions)

The Ternary Operator allows inline decision-making in a single concise line:

$$\text{Result} = \text{value\_if\_true} \quad \mathbf{if} \; (\text{condition}) \; \mathbf{else} \quad \text{value\_if\_false}$$

### Applications:
1. Finding the maximum / minimum of numbers.
2. Checking equality with inline fallbacks.
3. Nested ternary chains for 3-way status determinations.
4. One-line Palindrome verification.

```python
# Ternary Conditional Expressions Practical Suite

# 1. Maximum of two values with equality check
x, y = 45, 90
max_val = x if x > y else y if y > x else "Values are Equal"
print(f"Between {x} and {y}: Max = {max_val}")

# 2. Maximum among 3 numbers
a, b, c = 120, 450, 310
max_3 = a if (a >= b and a > c) else b if (b > a and b >= c) else c if (c >= a and c > b) else "All Equal"
print(f"Among {a}, {b}, {c}: Max = {max_3}")

# 3. Number sign classification (Positive / Negative / Zero)
val = -18
status = "Positive" if val > 0 else "Negative" if val < 0 else "Zero"
print(f"Number {val} is: {status}")

# 4. Palindrome string check in one line
word = "RACECAR"
is_palindrome = f"'{word}' is Palindrome" if word.lower() == word[::-1].lower() else f"'{word}' is NOT Palindrome"
print("Palindrome Drill:", is_palindrome)
```

## 6. Identity (`is`) vs. Equality (`==`) & Membership (`in`)

### Critical Distinction:
- **`==` (Equality)**: Compares **values** by invoking `__eq__()`. Returns `True` if two objects hold identical contents.
- **`is` (Identity)**: Compares **memory addresses** (`id(a) == id(b)`). Returns `True` ONLY if both variables point to the exact same object in RAM.
- **`in` (Membership)**: Tests whether a target element exists within an iterable sequence (string, list, tuple, set, dict keys).

```python
# Identity vs Equality & Membership Drills
# Equality vs Identity on Lists
list_1 = [1, 2, 3]
list_2 = [1, 2, 3]
list_3 = list_1

print(f"list_1 == list_2: {list_1 == list_2} (Same content)")
print(f"list_1 is list_2: {list_1 is list_2} (Different memory addresses: id1={id(list_1)}, id2={id(list_2)})")
print(f"list_1 is list_3: {list_1 is list_3} (Exact same reference)")

# Membership Testing
vowels = "aeiouAEIOU"
print(f"\n'e' in vowels       : {'e' in vowels}")
print(f"'x' not in vowels   : {'x' not in vowels}")

user_record = {"id": 101, "role": "admin", "active": True}
print(f"'role' in dict      : {'role' in user_record} (Checks dictionary KEYS)")
print(f"'admin' in dict     : {'admin' in user_record} (False: does not search values directly)")
print(f"'admin' in values() : {'admin' in user_record.values()}")
```

## 7. Assignment Expressions: The Walrus Operator (`:=`)

Introduced in Python 3.8, the **Walrus Operator (`:=`)** allows you to assign values to variables **inside an expression**.
It avoids redundant computations and simplifies `while` loops and list comprehensions.

```python
# Walrus Operator Demonstration
raw_data = "Software Development Engineer in Python"

# Traditional approach:
length = len(raw_data)
if length > 20:
    print(f"Traditional: Text is long ({length} characters)")

# Walrus approach: compute and assign inline!
if (n := len(raw_data)) > 20:
    print(f"Walrus (:=) inline: Text is long ({n} characters)")
```

## 8. Operator Precedence and Associativity Table

When multiple operators appear in a single expression, Python evaluates them based on predefined **Precedence** (order of priority) and **Associativity** (direction of evaluation: Left-to-Right or Right-to-Left).

| Priority | Operator | Description | Associativity |
|---|---|---|---|
| **1 (Highest)** | `()` | Parentheses (Grouping) | Left to Right |
| **2** | `**` | Exponentiation | **Right to Left** |
| **3** | `+x`, `-x`, `~x` | Unary positive, negative, bitwise NOT | Right to Left |
| **4** | `*`, `/`, `//`, `%` | Multiplications & Divisions | Left to Right |
| **5** | `+`, `-` | Additions & Subtractions | Left to Right |
| **6** | `<<`, `>>` | Bitwise Shifts | Left to Right |
| **7** | `&` | Bitwise AND | Left to Right |
| **8** | `^` | Bitwise XOR | Left to Right |
| **9** | `\|` | Bitwise OR | Left to Right |
| **10** | `==`, `!=`, `<`, `<=`, `>`, `>=`, `is`, `is not`, `in`, `not in` | Comparisons & Identity | Left to Right (Supports Chaining!) |
| **11** | `not` | Logical NOT | Right to Left |
| **12** | `and` | Logical AND | Left to Right |
| **13 (Lowest)**| `or` | Logical OR | Left to Right |

## 9. Short-Circuit Evaluation Mechanics

In Python, logical `and` and `or` do **NOT** return a boolean `True` or `False`.
Instead, they return the **exact operand value that determined the outcome**:

- **`X and Y`**:
  - If `X` is falsy, evaluates `X` and **immediately returns `X`** (short-circuit: `Y` is never touched).
  - If `X` is truthy, evaluates and returns `Y`.
- **`X or Y`**:
  - If `X` is truthy, evaluates `X` and **immediately returns `X`** (short-circuit: `Y` is never touched).
  - If `X` is falsy, evaluates and returns `Y`.

```python
# Short-Circuit Evaluation & Operand Return Demonstration
# 1. 'or' short-circuit: Returns first truthy value or last falsy value
print("'apple' or 'banana' :", repr('apple' or 'banana')) # 'apple'
print("'' or 'default'     :", repr('' or 'default'))     # 'default' (Classic default idiom!)
print("None or 0 or []     :", repr(None or 0 or []))     # [] (All falsy -> returns last)

# 2. 'and' short-circuit: Returns first falsy value or last truthy value
print("\n'admin' and 200    :", repr('admin' and 200))     # 200
print("0 and 'superuser'   :", repr(0 and 'superuser'))   # 0 (Short-circuits immediately!)
print("[] and (10 / 0)     :", repr([] and (10 / 0)))     # [] (Does NOT raise ZeroDivisionError!)
```

## 10. Production Pattern: Bitwise Flags & Security Bitmasks

Bitwise operators are universally used in production systems, operating systems, and network protocols to store and check multiple boolean flags inside a single lightweight integer:

- **Flag 1 (READ)**    : `0b001` (1)
- **Flag 2 (WRITE)**   : `0b010` (2)
- **Flag 3 (EXECUTE)** : `0b100` (4)

```python
# Bitwise Security Permissions System
READ    = 0b001  # 1
WRITE   = 0b010  # 2
EXECUTE = 0b100  # 4

# Assigning multiple permissions using Bitwise OR (|)
user_perms = READ | WRITE
print(f"User Permissions (READ | WRITE): {user_perms:03b} ({user_perms})")

# 1. Checking a permission using Bitwise AND (&)
has_read = bool(user_perms & READ)
has_exec = bool(user_perms & EXECUTE)
print(f"  Has READ permission?    : {has_read}")
print(f"  Has EXECUTE permission? : {has_exec}")

# 2. Granting a permission using Bitwise OR (|)
user_perms |= EXECUTE
print(f"  After granting EXECUTE  : {user_perms:03b} (Has EXECUTE: {bool(user_perms & EXECUTE)})")

# 3. Revoking a permission using Bitwise AND with NOT (& ~)
user_perms &= ~WRITE
print(f"  After revoking WRITE    : {user_perms:03b} (Has WRITE: {bool(user_perms & WRITE)})")

# 4. Toggling a permission using Bitwise XOR (^)
user_perms ^= READ  # Toggles READ off
print(f"  After toggling READ     : {user_perms:03b} (Has READ: {bool(user_perms & READ)})")
```

## 11. The Standard `operator` Module

Python includes the built-in `operator` module, which exports C-optimized functions for every built-in operator.
These are critical for functional pipelines, sorting by custom attributes, and performance-critical reductions.

```python
# Operator Module Demonstration
import operator
from functools import reduce

# Function equivalents of operators
print("operator.add(10, 20)  :", operator.add(10, 20))
print("operator.mul(6, 7)    :", operator.mul(6, 7))
print("operator.contains('Python', 'th'):", operator.contains('Python', 'th'))

# High-speed list multiplication reduction via operator.mul
nums = [1, 2, 3, 4, 5]
product = reduce(operator.mul, nums)
print(f"Product of {nums} via reduce(operator.mul) = {product}")

# High-speed itemgetter for sorting tabular data by specific columns
students = [("Chaitra", 88), ("Bhavna", 95), ("Aditya", 79)]
sorted_by_marks = sorted(students, key=operator.itemgetter(1), reverse=True)
print("Students sorted by marks (itemgetter):", sorted_by_marks)
```

## 12. Tricky Gotchas & Common Pitfalls

### ⚠️ Gotcha 1: Right-to-Left Associativity of Exponentiation (`**`)
Unlike multiplication, exponentiation evaluates from **right to left**:
`2 ** 3 ** 2` is evaluated as $2^{(3^2)} = 2^9 = 512$, **NOT** $(2^3)^2 = 8^2 = 64$!

```python
# Exponentiation Associativity Trap
expr_result = 2 ** 3 ** 2
print("2 ** 3 ** 2 evaluates to:", expr_result)
print("Equivalent to 2 ** (3 ** 2):", 2 ** (3 ** 2))
print("NOT equivalent to (2 ** 3) ** 2:", (2 ** 3) ** 2)
```

### ⚠️ Gotcha 2: Chained Comparison Surprises
Python evaluates chained comparisons like `a < b < c` as `(a < b) and (b < c)`.
However, tricky chains like `False == False in [False]` can cause major confusion!

```python
# Chained Comparison Gotchas
# What does (False == False in [False]) evaluate to?
# It translates to: (False == False) and (False in [False]) -> True and True -> True!
result = False == False in [False]
print("False == False in [False] evaluates to:", result)

# Contrast with explicit grouping:
print("(False == False) in [False]:", (False == False) in [False]) # True in [False] -> False!
```
