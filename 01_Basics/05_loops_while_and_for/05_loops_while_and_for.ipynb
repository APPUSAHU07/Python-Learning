## 1. Core Purpose & The 3 Golden Rules of Loop Design

> **Definition**: **Looping / Iterative / Repetitive Statements** execute a specific block of code repeatedly for a finite number of cycles until a termination condition becomes False.

### The 3 Mandatory Components of Every Loop:
Every robust loop must possess these three elements:
1. **Initialization**: Defining the starting state / counter variable.
2. **Condition / Test**: Evaluated prior to each iteration. If True, body executes; if False, loop terminates.
3. **Step / Update (Increment / Decrement)**: Modifying the state variable so it moves toward the termination condition, **preventing infinite loops**.

```
+--------------------------------------------------------------+
|                     LOOP LIFECYCLE FLOW                      |
|                                                              |
|        [ 1. Initialization: i = 1 ]                          |
|                     |                                        |
|                     v                                        |
|        [ 2. Condition: i <= 10 ]  ----(False)----> [ Exit ]  |
|                     |                                        |
|                  (True)                                      |
|                     v                                        |
|             { Loop Body Code }                               |
|                     |                                        |
|                     v                                        |
|        [ 3. Update Step: i += 1 ]                            |
|                     |                                        |
|                     +------------> Loops back to Step 2      |
+--------------------------------------------------------------+
```

## 2. `while` Loop vs. `for` Loop Paradigms

| Feature | `while` Loop | `for` Loop |
|---|---|---|
| **Primary Philosophy** | **Condition-driven** | **Collection / Sequence-driven** |
| **Best Used When** | Number of iterations is **not known in advance** (e.g., polling, user input, math reductions). | Iterating over a **known sequence, range, or iterable collection**. |
| **State Variable** | Managed manually by programmer (`i += 1`). | Managed automatically by Python iterator protocol. |
| **Risk of Infinite Loop** | High (if update step is omitted). | Extremely low (terminates when iterable exhausts). |

````python
# while vs for Loop Comparison
# Problem: Sum of first N natural numbers (N = 10)
N = 10

# 1. while loop approach (Explicit Initialization, Condition, Update)
i = 1
sum_while = 0
while i <= N:
    sum_while += i
    i += 1  # Mandatory step

# 2. for loop approach (Automatic sequence iteration via range)
sum_for = sum(range(1, N + 1))

print(f"Sum of first {N} natural numbers:")
print(f"  via while loop = {sum_while}")
print(f"  via for loop   = {sum_for}")
print(f"Both produce identical results: {sum_while == sum_for}")
````

## 3. The Python Loop `else` Clause

> **A Unique Python Architectural Feature**: Both `while` and `for` loops support an optional `else` block!
>
> **The Golden Rule of Loop `else`**:
> The `else` block executes **ONLY if the loop completes all iterations normally** (i.e., condition becomes False or sequence exhausts).
> If the loop is terminated prematurely by a **`break` statement**, the `else` block is **COMPLETELY SKIPPED**.

````python
# The Loop else Clause: Search & Early Exit Demonstration
def search_student_id(target_id, database):
    print(f"Searching for Student ID {target_id} in {database}...")
    for idx, sid in enumerate(database):
        if sid == target_id:
            print(f"  FOUND: Student {target_id} located at index {idx}!")
            break  # Triggers early exit -> SKIPS the 'else' block!
    else:
        # Executes ONLY if NO 'break' was encountered!
        print(f"  NOT FOUND: Student {target_id} is not present in records.")

records = [101, 102, 105, 108, 112]
search_student_id(105, records)  # Found -> break triggers -> skips else
search_student_id(999, records)  # Not found -> loop finishes -> executes else
````

## 4. Inner / Nested Loops Architecture

> **Definition**: Defining one loop inside the body of another loop creates an **Inner / Nested Loop**.
>
> **Execution Rule**:
> **For every single cycle of the Outer Loop, the Inner Loop executes its entire cycle from start to finish.**
> If the outer loop runs $M$ times and the inner loop runs $N$ times, the inner body executes $M \times N$ times ($O(M \times N)$ time complexity).

````python
# Nested Loops: Generating Multiplication Tables
start_table = 2
end_table = 4

print("=" * 45)
print("     NESTED MULTIPLICATION TABLES")
print("=" * 45)

for t in range(start_table, end_table + 1):  # Outer Loop
    print(f"--- Multiplication Table for {t} ---")
    for step in range(1, 6):                # Inner Loop (First 5 multiples)
        print(f"  {t} x {step} = {t * step:>2d}")
    print()
print("=" * 45)
````

## 5. Prime Number Algorithms & Optimization

A positive integer $n > 1$ is prime if it has no positive divisors other than 1 and itself.
- **Naive Check**: Check divisibility from $2$ to $n-1$ ($O(n)$).
- **Optimized Check**: Check divisibility up to $\sqrt{n}$ ($O(\sqrt{n})$) because if $n = a \times b$, one factor must be $\le \sqrt{n}$.

````python
# Prime Number Testing & Range Generator using Loop-else
def is_prime(num):
    if num <= 1:
        return False
    for factor in range(2, int(num ** 0.5) + 1):
        if num % factor == 0:
            return False  # Composite
    return True

# Prime Numbers within range [10, 50] using nested loop & loop-else
low, high = 10, 50
primes_found = []

for val in range(low, high + 1):
    if val > 1:
        for factor in range(2, int(val ** 0.5) + 1):
            if val % factor == 0:
                break
        else:
            primes_found.append(val)

print(f"Prime numbers in range [{low}, {high}]:")
print(f"  {primes_found}")
print(f"Total primes found: {len(primes_found)}")
````

## 6. Transfer Flow Statements: `break`, `continue`, `pass`

| Statement | Action | Typical Use Case |
|---|---|---|
| `break` | Immediately terminates the enclosing loop; jumps to first statement outside. | Early exit on target found or error detected. |
| `continue` | Aborts current iteration; skips remaining statements; jumps to next cycle. | Filtering unwanted data, skipping whitespace/vowels. |
| `pass` | No-operation (no-op). Syntactic placeholder for empty blocks. | Stubbing empty functions, classes, or exception blocks. |

````python
# Transfer Flow Statements Demonstration
# 1. continue: Skipping specific values (Filter out vowels and spaces)
raw_text = "PYTHON PROGRAMMING 3.14"
consonants_only = []

for char in raw_text:
    if char.upper() in "AEIOU 0123456789.":
        continue  # Skip vowels, numbers, spaces
    consonants_only.append(char)

print("Original Text      :", raw_text)
print("Consonants Extracted:", "".join(consonants_only))

# 2. pass: Syntactic stub
def future_feature_placeholder():
    pass  # Prevents IndentationError on empty body

print("Pass statement executed cleanly without syntax error.")
````

## 7. Numerical Algorithms: Digits Sum & Student Marks Reporting

### Algorithm 1: Sum of Digits of a Number
Extracts digits from right to left using modulo (`% 10`) and removes them using floor division (`// 10`).

````python
# Sum of Digits Algorithm (while loop)
number = 94827
temp = number
digit_sum = 0
digits_reversed = []

while temp > 0:
    digit = temp % 10          # Extract last digit
    digit_sum += digit         # Accumulate sum
    digits_reversed.append(str(digit))
    temp = temp // 10          # Discard last digit

print(f"Number              : {number}")
print(f"Extracted Digits    : {' + '.join(digits_reversed)} = {digit_sum}")
print(f"Arithmetically Rev  : {''.join(digits_reversed)}")
````

### Algorithm 2: Student Marks Report with Validation & Grading
Computes total, percentage, subject-level pass validation (marks $\ge 40$), and classification (Distinction $\ge 75\%$, First Class $\ge 60\%$, Second Class $\ge 50\%$).

````python
# Student Marks Report Generation Engine
def generate_student_report(sno, name, marks_dict):
    marks = list(marks_dict.values())
    total_marks = sum(marks)
    max_marks = len(marks) * 100
    percentage = (total_marks / max_marks) * 100.0
    
    # Rule: Student passes ONLY if every subject score >= 40
    passed = all(score >= 40 for score in marks)
    
    if not passed:
        grade = "FAIL"
    elif percentage >= 75:
        grade = "DISTINCTION"
    elif percentage >= 60:
        grade = "FIRST CLASS"
    elif percentage >= 50:
        grade = "SECOND CLASS"
    else:
        grade = "PASS CLASS"

    print("=" * 55)
    print("               ACADEMIC MARKS REPORT")
    print("=" * 55)
    print(f"  Roll Number  : {sno}")
    print(f"  Student Name : {name}")
    print("-" * 55)
    for subject, score in marks_dict.items():
        status = "Pass" if score >= 40 else "FAIL"
        print(f"  {subject:<20}: {score:>3d} / 100  [{status}]")
    print("-" * 55)
    print(f"  Total Score  : {total_marks} / {max_marks}")
    print(f"  Percentage   : {percentage:.2f}%")
    print(f"  Final Status : {grade}")
    print("=" * 55)

generate_student_report(101, "Suresh Kumar", {
    "Mathematics": 88,
    "Data Structures": 92,
    "Python Programming": 95,
    "Computer Networks": 78,
    "Operating Systems": 84
})
````

## 8. Pythonic Iteration Utilities: `enumerate`, `zip`, and `reversed`

Idiomatic Python avoids manual index bookkeeping (`for i in range(len(lst)):`) in favor of expressive built-in iteration helpers:

1. **`enumerate(iterable, [start=0])`**: Returns pairs of `(index, item)`. Eliminates manual `index += 1`.
2. **`zip(*iterables)`**: Pairs elements from multiple collections in lockstep. Terminates on the shortest iterable.
3. **`itertools.zip_longest(*iterables, fillvalue=None)`**: Pairs elements up to the longest collection, filling gaps.
4. **`reversed(sequence)`**: Returns an iterator traversing backwards without copying the collection.

````python
# Pythonic Iteration Suite: enumerate, zip, and reversed
students = ["Aditya", "Bhavna", "Charan", "Divya"]
courses = ["Python Full Stack", "Data Engineering", "Cloud DevOps", "AI & ML"]
scores = [92, 88, 95, 84]

# 1. enumerate with custom 1-based index
print("--- Student Enrollment Listing (enumerate) ---")
for rank, name in enumerate(students, start=1):
    print(f"  Rank #{rank}: {name}")

# 2. zip across 3 collections simultaneously
print("\n--- Unified Student Profile (zip) ---")
for name, course, score in zip(students, courses, scores):
    print(f"  {name:<10} | Course: {course:<18} | Score: {score}")

# 3. reversed lazy traversal
print("\n--- Reversed Sequence Traversal ---")
for item in reversed(students):
    print(f"  {item}", end=" ")
print()
````

## 9. Classic Number Theory Algorithms (Loops Mastery)

### Algorithm 1: Euclidean Algorithm for Greatest Common Divisor (GCD)
Computes the largest common divisor of two integers using remainder division:

$$\text{GCD}(a, b) = \text{GCD}(b, a \% b) \quad \text{until } b = 0$$

### Algorithm 2: Armstrong (Narcissistic) Number
A number that equals the sum of its own digits each raised to the power of the total number of digits:

$$153 = 1^3 + 5^3 + 3^3 = 1 + 125 + 27 = 153$$

````python
# GCD, LCM, and Armstrong Number Verification Suite
def compute_gcd(a, b):
    """Euclidean Algorithm for GCD."""
    while b != 0:
        a, b = b, a % b
    return a

def compute_lcm(a, b):
    """LCM via mathematical identity: LCM(a, b) = |a * b| / GCD(a, b)"""
    return abs(a * b) // compute_gcd(a, b)

val1, val2 = 48, 18
print(f"GCD of ({val1}, {val2}) = {compute_gcd(val1, val2)}")
print(f"LCM of ({val1}, {val2}) = {compute_lcm(val1, val2)}")

# Armstrong Number Verification
def is_armstrong(num):
    s = str(num)
    k = len(s)
    total = sum(int(ch) ** k for ch in s)
    return total == num

print("\n--- Armstrong Numbers in Range [1, 1000] ---")
armstrong_numbers = [n for n in range(1, 1001) if is_armstrong(n)]
print("Found Armstrong Numbers:", armstrong_numbers)
````

## 10. Tricky Gotchas & Common Pitfalls

### ⚠️ Gotcha 1: Modifying a List While Iterating Over It
Modifying a list in-place (e.g. `lst.remove(item)`) while iterating with a `for` loop alters internal index positions and silently skips elements!

````python
# Modifying List During Iteration Trap
numbers = [1, 2, 2, 3, 4]

# TRAP (Buggy!):
for x in numbers:
    if x == 2:
        numbers.remove(x)  # Skips the second '2'!

print("Buggy removal result:", numbers)  # Still contains [1, 2, 3, 4]!

# CORRECT PRODUCTION FIX: Use list comprehension or copy!
correct_numbers = [x for x in [1, 2, 2, 3, 4] if x != 2]
print("Correct removal result:", correct_numbers)
````

### ⚠️ Gotcha 2: The Infinite Loop Omission
Omitting the update step in a `while` loop causes an infinite loop that freezes CPU threads.

````python
# Safe While Loop with Safety Breaker
counter = 0
while True:
    counter += 1
    if counter >= 5:
        print(f"Safety breaker triggered at counter = {counter}. Exiting infinite loop safely.")
        break
````

## 11. Summary & Quick Revision Cheat Sheet

| Loop Construct | Primary Purpose | How to Terminate | Important Nuance |
|---|---|---|---|
| **`while`** | Condition-driven iteration | When condition becomes False or via `break`. | Must manually update counter variable. |
| **`for`** | Sequence / range iteration | When sequence exhausts or via `break`. | Fast C-level iteration; no index bookkeeping needed. |
| **`break`** | Early exit from loop | Immediately exits innermost loop. | **Skips** the loop's `else` block! |
| **`continue`** | Skip remainder of cycle | Jumps directly to next cycle. | Preserves loop continuation. |
| **`else` on Loop** | Execution on completion | Runs ONLY if loop completed without `break`. | Perfect for search loops and prime checks. |
