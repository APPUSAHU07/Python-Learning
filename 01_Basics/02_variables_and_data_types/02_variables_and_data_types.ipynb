## 1. Core Purpose of Data Types

> **Definition**: In Python, the purpose of a **DataType** is to allocate sufficient, appropriate memory space in the computer's RAM where inputs (literals) can be stored and processed safely.
>
> In Python, **data types are classes**, and variables holding values are **instances (objects)** of these classes.
>
> Python provides **14 built-in data types**, systematically classified into **6 categories**:

| Category | Category Purpose | Data Types Included | Mutability Status |
|---|---|---|---|
| **I. Fundamental** | Stores a single atomic value | `int`, `float`, `bool`, `complex` | **All Immutable** |
| **II. Sequence** | Stores ordered sequence of items | `str`, `bytes`, `bytearray`, `range` | `bytearray` is **Mutable**; `str`, `bytes`, `range` are **Immutable** |
| **III. List Category** | Ordered collection of elements | `list`, `tuple` | `list` is **Mutable**; `tuple` is **Immutable** |
| **IV. Set Category** | Unordered collection of unique items | `set`, `frozenset` | `set` is **Mutable**; `frozenset` is **Immutable** |
| **V. Dict Category** | Key-Value associative mapping | `dict` | **Mutable** |
| **VI. None Type** | Represents absence of value / singleton | `NoneType` (`None`) | **Immutable** |

## 2. Identifiers, Variables & Naming Rules

### What is an Identifier?
> **Definition**: When input literals are stored in main memory, programmers must give distinct, human-readable names to identify those memory locations. These names are called **Identifiers** or **Variable Names**.

### The 6 Golden Rules of Identifiers:
1. **Allowed Characters**: Uppercase letters (`A-Z`), lowercase letters (`a-z`), digits (`0-9`), and the underscore (`_`).
2. **First Character Rule**: Must begin with a letter (`a-z`, `A-Z`) or an underscore (`_`). **Cannot start with a digit** (e.g., `1total` is invalid, `total_1` is valid).
3. **No Special Symbols**: Characters like `@`, `$`, `#`, `%`, `&`, `!` are strictly disallowed.
4. **Case-Sensitivity**: Python is strictly case-sensitive. `marks`, `Marks`, and `MARKS` reference three completely independent memory addresses.
5. **No Reserved Keywords**: Python's 35 built-in keywords (such as `def`, `class`, `if`, `while`, `return`) cannot be used as identifier names.
6. **Recommended Convention**: Use descriptive `snake_case` for variables/functions (PEP 8 standard).

````python
# Inspecting Python Reserved Keywords & Identifier Validation
import keyword

print("Total Python Keywords:", len(keyword.kwlist))
print("-" * 55)
# Print keywords neatly formatted in rows of 5
for i in range(0, len(keyword.kwlist), 5):
    print("  ".join(f"{k:<10}" for k in keyword.kwlist[i:i+5]))
print("-" * 55)

# Identifier inspection using isidentifier()
test_identifiers = ["student_name", "_emp_id", "2nd_rank", "for", "total$", "validVar123"]
print("Identifier Validity Check:")
for name in test_identifiers:
    valid = name.isidentifier() and not keyword.iskeyword(name)
    print(f"  {name:<15} -> {'VALID' if valid else 'INVALID'}")
````

## 3. Literals & Base Number Representation

Literals are the raw data values passed into a Python program. Integer values can be represented in **4 number systems**:

| Number System | Base | Allowed Digits | Python Prefix | Conversion Function |
|---|---|---|---|---|
| **Decimal** | Base 10 | `0` to `9` | None (Default) | `int()` |
| **Binary** | Base 2 | `0`, `1` | `0b` or `0B` | `bin()` |
| **Octal** | Base 8 | `0` to `7` | `0o` or `0O` | `oct()` |
| **Hexadecimal** | Base 16 | `0` to `9`, `A` to `F` (or `a-f`) | `0x` or `0X` | `hex()` |

````python
# Base Number Representations and Bidirectional Conversions
decimal_val = 45

# Converting decimal to binary, octal, hexadecimal
b_val = bin(decimal_val)
o_val = oct(decimal_val)
h_val = hex(decimal_val)

print(f"Decimal Value       : {decimal_val}")
print(f"Binary Representation : {b_val} (Type: {type(b_val).__name__})")
print(f"Octal Representation  : {o_val} (Type: {type(o_val).__name__})")
print(f"Hex Representation    : {h_val} (Type: {type(h_val).__name__})")

# Direct literal assignment using prefixes
num_bin = 0b101101   # 45 in binary
num_oct = 0o55       # 45 in octal
num_hex = 0x2D       # 45 in hex

print(f"\nEvaluating prefixed literals directly in Python:")
print(f"  0b101101 = {num_bin}")
print(f"  0o55     = {num_oct}")
print(f"  0x2D     = {num_hex}")
print(f"All evaluate to the exact same integer value: {num_bin == num_oct == num_hex == 45}")
````

## 4. Fundamental Category Data Types: `int`, `float`, `bool`, `complex`

### Properties of Fundamental Types:
- Stores **single, atomic values**.
- Fixed mathematical semantics and **immutable** memory behavior.
- In Python 3, `int` has **arbitrary precision** (limited only by available RAM, no 32-bit or 64-bit overflow!).
- `complex` numbers are represented as `real + imag*j`, where `j = √(-1)`.

````python
# Fundamental Data Types Drill
# 1. Arbitrary Precision Integer (No overflow!)
huge_int = 10 ** 40
print(f"Huge Integer (10^40):\n  {huge_int}")
print(f"Byte size of huge_int: {sys.getsizeof(huge_int)} bytes")

# 2. Floating-point scientific notation
sci_float = 2.45e4   # 2.45 * 10^4 = 24500.0
small_float = 1.2e-3 # 0.0012
print(f"\nScientific floats: {sci_float}, {small_float}")

# 3. Complex numbers
comp_num = 3.5 + 4.2j
print(f"\nComplex Number : {comp_num}")
print(f"  Real Part    : {comp_num.real}")
print(f"  Imag Part    : {comp_num.imag}")
print(f"  Conjugate    : {comp_num.conjugate()}")
````

## 5. Sequence Category: `str`, `bytes`, `bytearray`, `range`

### Deep Dive: `bytes` vs. `bytearray` (End-to-End Encryption Sequence)
> **Purpose**: Used for high-speed network communication, binary file protocols, image buffers, and End-to-End Encryption (E2EE).
> **Value Constraint**: Each element must be a numerical integer in the range `0 <= x <= 255` (`(0, 256)`).
> - `bytes`: **Immutable**. Once created, elements cannot be modified.
> - `bytearray`: **Mutable**. Allows in-place item assignment (`ba[0] = 90`).

````python
# Bytes vs Bytearray Operations
# Initializing byte buffer from integer sequence
raw_ints = [72, 101, 108, 108, 111]  # ASCII for 'Hello'

b_immutable = bytes(raw_ints)
ba_mutable = bytearray(raw_ints)

print("Original bytes object     :", b_immutable, f"| Decoded: {b_immutable.decode('ascii')}")
print("Original bytearray object :", ba_mutable, f"| Decoded: {ba_mutable.decode('ascii')}")

# Mutating bytearray in-place
ba_mutable[0] = 74   # Change 'H' (72) to 'J' (74) -> 'Jello'
ba_mutable.append(33) # Append '!' (33)
print("Mutated bytearray in-place:", ba_mutable, f"| Decoded: {ba_mutable.decode('ascii')}")

# Attempting item assignment on bytes raises TypeError:
try:
    b_immutable[0] = 74
except TypeError as err:
    print(f"Expected Error on bytes: {err}")
````

### Deep Dive: `range` Data Type
> **Purpose**: Generates an immutable arithmetic sequence of integers with equal spacing.
> **Syntax 1**: `range(stop)` -> starts at `0`, step `+1`, ends at `stop - 1`.
> **Syntax 2**: `range(start, stop)` -> starts at `start`, step `+1`, ends at `stop - 1`.
> **Syntax 3**: `range(start, stop, step)` -> starts at `start`, step `step`, ends before `stop`.
> 
> **Memory Efficiency**: Storing `range(1_000_000)` consumes only 48 bytes of RAM because it computes values on-the-fly!

````python
# Range Syntaxes and Slicing
r1 = range(5)             # 0, 1, 2, 3, 4
r2 = range(10, 15)        # 10, 11, 12, 13, 14
r3 = range(50, 20, -5)    # 50, 45, 40, 35, 30, 25 (Negative step)

print("r1 list representation:", list(r1))
print("r2 list representation:", list(r2))
print("r3 list representation:", list(r3))

# Range supports indexing and slicing!
print("r3[0] (First element)  :", r3[0])
print("r3[-1] (Last element)  :", r3[-1])
print("r3[1:4] (Slicing)      :", list(r3[1:4]))
print(f"Memory size of range(10,000,000): {sys.getsizeof(range(10_000_000))} bytes (Constant O(1) Memory)")
````

## 6. Comprehensive Type-Casting Matrix

> **Type Casting**: The process of converting a value of one data type into another possible data type.
> Python provides **5 fundamental type-casting functions**:

```
+-------------------------------------------------------------+
|                THE 5 TYPE-CASTING FUNCTIONS                 |
|                                                             |
|   1. int()     --> Converts to integer                      |
|   2. float()   --> Converts to floating-point number        |
|   3. bool()    --> Converts to boolean (True / False)       |
|   4. complex() --> Converts to complex number (a + bj)      |
|   5. str()     --> Converts to human-readable string        |
+-------------------------------------------------------------+
```

### The Golden Rule of Boolean Truthiness (`bool()`):
- **False Values**: Numeric Zero (`0`, `0.0`, `0j`), Singleton `None`, and all empty collections (`""`, `[]`, `()`, `{}`, `set()`, `range(0)`).
- **True Values**: **All non-zero numerical values** (positive and negative: `10`, `-5`, `0.0001`) and all non-empty collections.

````python
# Exhaustive Type-Casting Matrix Implementation
print("=" * 50)
print("             TYPE CASTING DRILLS")
print("=" * 50)

# 1. int() Conversions
print(f"int(12.85)             -> {int(12.85)}")
print(f"int(True)              -> {int(True)}")
print(f"int(False)             -> {int(False)}")
print(f"int('1024')            -> {int('1024')}")
print(f"int('0b101', 2)        -> {int('0b101', 2)} (Binary string to int)")
print(f"int('0xFF', 16)        -> {int('0xFF', 16)} (Hex string to int)")

# 2. float() Conversions
print(f"\nfloat(50)             -> {float(50)}")
print(f"float(True)            -> {float(True)}")
print(f"float('3.14159')       -> {float('3.14159')}")

# 3. bool() Conversions (Truthiness Matrix)
print(f"\nbool(100)              -> {bool(100)}")
print(f"bool(-25)              -> {bool(-25)} (Negative is True!)")
print(f"bool(0)                -> {bool(0)}")
print(f"bool('False')          -> {bool('False')} (Non-empty string is True!)")
print(f"bool('')               -> {bool('')} (Empty string is False)")
print(f"bool([0])              -> {bool([0])} (Non-empty list is True)")
print(f"bool([])               -> {bool([])} (Empty list is False)")

# 4. complex() Conversions
print(f"\ncomplex(10)            -> {complex(10)}")
print(f"complex(5, -8)          -> {complex(5, -8)}")
print(f"complex(True, False)   -> {complex(True, False)}")

# 5. str() Conversions
print(f"\nstr(999)               -> {repr(str(999))}")
print(f"str([1, 2, 3])         -> {repr(str([1, 2, 3]))}")
print("=" * 50)
````

## 7. Mutability vs. Immutability Architecture

> **Mutable Object**: An object that permits in-place modifications at the **same memory address (`id()`)**.
> **Immutable Object**: An object whose value cannot be altered in-place. Any modification creates a **new object at a new memory address**.

```
IMMUTABLE BEHAVIOR (int):
Step 1: x = 100 ---------> [ Memory Address: 0x1000 | Value: 100 ]
Step 2: x = x + 1 -------> [ Memory Address: 0x2000 | Value: 101 ]  (NEW ADDRESS!)

MUTABLE BEHAVIOR (list):
Step 1: lst = [10, 20] --> [ Memory Address: 0x5000 | Items: [10, 20] ]
Step 2: lst.append(30) --> [ Memory Address: 0x5000 | Items: [10, 20, 30] ] (SAME ADDRESS!)
```

````python
# Memory Verification of Mutability vs Immutability
# Immutable Demo: Integer
val = 500
id_before = id(val)
val += 1
id_after = id(val)

print("--- IMMUTABLE INTEGER ---")
print(f"Address Before: {id_before}")
print(f"Address After : {id_after}")
print(f"Memory Address Changed? {id_before != id_after}")

# Mutable Demo: List
arr = [1, 2, 3]
id_list_before = id(arr)
arr.append(4)
id_list_after = id(arr)

print("\n--- MUTABLE LIST ---")
print(f"Address Before: {id_list_before}")
print(f"Address After : {id_list_after}")
print(f"Memory Address Preserved? {id_list_before == id_list_after}")
````

## 8. CPython Memory Internals: Variables as Named Pointers (`PyObject`)

### Contrast: C-Style Variables vs. Python Named References:
- **In C / C++ (Memory Boxes)**: Declaring `int x = 10;` allocates a fixed 4-byte box in RAM. The variable `x` *is* the box. Changing `x = 20` overwrites the contents inside that same box.
- **In Python (Object References)**: Executing `x = 10` allocates a full **`PyObject` heap structure** for the integer `10`. The variable `x` is simply a **named reference (pointer)** pointing to that heap object!

```
C-Style (Box Architecture):
+--------------+
| x: [  10  ]  |  <--- Overwritten in-place when modified
+--------------+

Python (Object Pointer Architecture):
Variable Name:           Heap-Allocated PyObject Structure:
+-------+                +-------------------------------+
|   x   | -------------> | ob_refcnt : 1                 |
+-------+                | ob_type   : <class 'int'>     |
                         | ob_digit  : 10                |
                         +-------------------------------+
```

````python
# Memory Reference Inspection via sys.getsizeof and ctypes
import sys
import ctypes

num = 42

# Memory size of a simple Python integer vs C 4-byte int
print(f"Value               : {num}")
print(f"Memory address      : {hex(id(num))}")
print(f"Total PyObject Size : {sys.getsizeof(num)} bytes (28 bytes overhead for C struct headers!)")

# Inspecting reference count directly from C struct header
ref_count = ctypes.c_long.from_address(id(num)).value
print(f"Active C References : {ref_count}")
````

## 9. Hashability & Immutable Type Safety (`hash()`)

> **Definition**: An object is **Hashable** if it has a hash value that **remains constant throughout its entire lifetime** and can be compared to other objects (`__eq__`).
>
> **The Golden Rule of Hashability**:
> - **Immutable Types ARE Hashable**: `int`, `float`, `str`, `tuple`, `frozenset`, `bytes`.
>   *(Can be safely used as Dictionary Keys and Set Elements).*
> - **Mutable Types ARE NOT Hashable**: `list`, `dict`, `set`, `bytearray`.
>   *(Attempting to use them as dict keys or in sets raises `TypeError: unhashable type`).*

````python
# Hashability Verification & Dictionary Key Safety
# 1. Valid hashable dictionary keys
valid_cache = {
    ("127.0.0.1", 8080): "Local Web Server",  # Immutable Tuple Key
    "user_session": "active_token_99",        # Immutable String Key
    frozenset({1, 2, 3}): "Config Set",       # Immutable Frozenset Key
    42: "Answer to Everything"                # Immutable Integer Key
}

print("Dictionary with Hashable Keys initialized successfully:")
for k, v in valid_cache.items():
    print(f"  Key ({type(k).__name__:<9}) [Hash: {hash(k):>20}] -> {v}")

# 2. Attempting to use a mutable list as a dictionary key or set item
try:
    bad_dict = {[1, 2]: "Invalid List Key"}
except TypeError as err:
    print(f"\nExpected Hash Failure on Mutable List: {err}")
````

## 10. Boolean Arithmetic & Subclass Behavior

In Python, `bool` is explicitly implemented as a **subclass of `int`**:
- `issubclass(bool, int) == True`
- `True` has an integer value of `1`.
- `False` has an integer value of `0`.

This enables high-performance **branchless programming** (e.g. counting matches without `if` statements).

````python
# Boolean Arithmetic & Branchless Counting
print("Is bool a subclass of int?", issubclass(bool, int))
print("True + True + True       =", True + True + True)
print("False * 50               =", False * 50)
print("True == 1                =", True == 1)

# Branchless counting: Count students with passing grades (>= 50) without 'if'
test_scores = [78, 45, 92, 38, 65, 88, 49, 90]
passed_count = sum(score >= 50 for score in test_scores)
print(f"\nTest scores: {test_scores}")
print(f"Total students who passed (branchless sum): {passed_count} / {len(test_scores)}")
````

## 11. Tricky Gotchas & Common Pitfalls

### ⚠️ Gotcha 1: Direct Float String to Integer Casting
Calling `int('12.34')` directly raises a `ValueError` because `int()` expects integer literal syntax when parsing strings.
To safely cast a floating-point string to an integer, chain: `int(float('12.34'))`.

````python
# Safe Float String Casting
num_str = "12.75"

try:
    bad_cast = int(num_str)
except ValueError as err:
    print(f"Direct int('12.75') FAILED: {err}")

# Safe two-step conversion:
good_cast = int(float(num_str))
print(f"Safe chained int(float('12.75')): {good_cast} (Truncates decimal to 12)")
````

### ⚠️ Gotcha 2: Floating Point Precision Inaccuracy
Computers store floating-point numbers in binary (IEEE 754 standard). Certain decimal fractions (like `0.1`) cannot be represented with finite binary digits, causing rounding discrepancies.

````python
# IEEE 754 Floating-Point Precision Trap
calc = 0.1 + 0.2
print(f"0.1 + 0.2 evaluates to: {calc:.17f}")
print(f"0.1 + 0.2 == 0.3? -> {calc == 0.3} (Surprise!)")

# Correct Production Comparison using math.isclose() or decimal module
import math
print(f"math.isclose(0.1 + 0.2, 0.3)? -> {math.isclose(calc, 0.3)}")

from decimal import Decimal
dec_calc = Decimal('0.1') + Decimal('0.2')
print(f"Decimal('0.1') + Decimal('0.2') == Decimal('0.3')? -> {dec_calc == Decimal('0.3')}")
````
