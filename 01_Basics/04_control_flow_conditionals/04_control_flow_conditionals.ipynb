## 1. Core Purpose of Conditional Branching Statements

> **Definition**: By default, the Python Virtual Machine (PVM) executes instructions sequentially from top to bottom.
> **Conditional / Selection / Branching Statements** alter this sequential flow by evaluating a boolean expression to execute either **Block X (if True)** or **Block Y (if False)** exactly once.

### The 4 Fundamental Types of Conditional Statements:

| Statement Type | Execution Logic | Syntax Pattern | Recommended Use Case |
|---|---|---|---|
| **1. Simple `if`** | Executes block ONLY if condition is True; does nothing if False. | `if condition:` | Standalone guard checks, validation filters. |
| **2. `if..else`** | Executes Block-1 if True, or alternative Block-2 if False. | `if condition: ... else:` | Binary decisions (Even/Odd, Pass/Fail). |
| **3. `if..elif..else`** | Evaluates multiple mutually exclusive conditions sequentially. | `if: ... elif: ... else:` | Multi-tier categorization (Grades, Taxes, Age brackets). |
| **4. `match..case`** | Structural pattern matching (Python 3.10+). | `match val: case ...:` | Menu-driven dispatching, command parsers, AST matching. |

## 2. Indentation & Block Architecture in Python

Unlike C, C++, or Java which delimit code blocks using curly braces `{ ... }`, **Python uses significant indentation**:
- Every block is introduced with a colon `:`.
- All statements inside that block must share the exact same indentation level.
- **PEP 8 Standard**: Always use **4 spaces per indentation level** (never mix tabs and spaces).

```
if condition_1:
    # Level 1 Indentation (4 spaces)
    statement_1
    if condition_2:
        # Level 2 Indentation (8 spaces)
        statement_2
statement_outside  # Level 0 (Normal sequential execution resumes)
```

````python
# Simple if and if-else Architectural Demonstrations
# 1. Simple if: Guard condition
age = 20
if age >= 18:
    print(f"Age {age}: Eligible to vote in national elections.")

# 2. if-else: Binary decision (Even / Odd via two different methods)
number = 47

# Method A: Modulo operator (% 2)
if number % 2 == 0:
    res_a = "Even"
else:
    res_a = "Odd"

# Method B: Bitwise AND (& 1) - Highly optimized!
# (Odd numbers always have lowest binary bit = 1; Even numbers have 0)
if number & 1 == 0:
    res_b = "Even"
else:
    res_b = "Odd"

print(f"Number {number}: Modulo Result = {res_a} | Bitwise Result = {res_b}")
````

## 3. Digit-to-Word Conversion: 3 Comparative Paradigms

A classic software design drill showing how the same real-world problem is solved across beginner, intermediate, and advanced Python paradigms:

1. **Approach 1**: Dictionary Lookup with `dict.get()` and if-else validation.
2. **Approach 2**: Inline Ternary Expression with dictionary safety check.
3. **Approach 3**: Direct inline fallback retrieval with `dict.get(key, default)`.

````python
# Digit-to-Word Conversion: 3 Approaches
digits_map = {
    0: "ZERO", 1: "ONE", 2: "TWO", 3: "THREE", 4: "FOUR",
    5: "FIVE", 6: "SIX", 7: "SEVEN", 8: "EIGHT", 9: "NINE"
}

test_digit = 8

# Approach 1: Dictionary get() with if-else validation
res1 = digits_map.get(test_digit)
if res1 is not None:
    print(f"Approach 1 (if-else validation) : {test_digit} is {res1}")
else:
    print(f"Approach 1 : {test_digit} is a multi-digit number or invalid")

# Approach 2: Ternary Expression with dict lookup
res2 = digits_map.get(test_digit) if digits_map.get(test_digit) is not None else "Not a single digit"
print(f"Approach 2 (Ternary Expression)   : {test_digit} is {res2}")

# Approach 3: Pythonic inline default retrieval
print(f"Approach 3 (dict.get with default): {test_digit} -> {digits_map.get(test_digit, 'Multi-Digit Number')}")
````

## 4. Multi-Branch Real-World Pricing Engine

Demonstrating cascading `if-elif-else` logic for cinema ticket pricing with tiered age categories, day-of-week promotional discounts, and edge-case validation.

````python
# Cinema Ticket Pricing Engine
def calculate_cinema_ticket(customer_name, age, day_of_week):
    if age < 0 or age > 120:
        return f"Error: Invalid age '{age}' specified!"
    
    # Base price based on age tier
    if age < 5:
        base_price = 0.0
        tier = "Infant (Free Admission)"
    elif age <= 12:
        base_price = 100.0
        tier = "Child"
    elif age <= 60:
        base_price = 220.0
        tier = "Adult"
    else:
        base_price = 140.0
        tier = "Senior Citizen (Concession)"

    # Promotional Discount: 25% off on Tuesdays and Wednesdays
    discount = 0.0
    if day_of_week.strip().lower() in ["tuesday", "wednesday"] and base_price > 0:
        discount = base_price * 0.25
    
    final_price = base_price - discount
    return {
        "customer": customer_name,
        "age": age,
        "tier": tier,
        "base_price": base_price,
        "discount": discount,
        "final_price": final_price
    }

# Running test scenarios
scenarios = [
    ("Aarav", 4, "Sunday"),
    ("Diya", 10, "Wednesday"),
    ("Rajesh", 35, "Wednesday"),
    ("Kavita", 68, "Friday")
]

print("=" * 65)
print(f"{'NAME':<10} | {'AGE':<4} | {'TIER':<25} | {'FINAL PRICE':<12}")
print("=" * 65)
for name, a, day in scenarios:
    res = calculate_cinema_ticket(name, a, day)
    print(f"{res['customer']:<10} | {res['age']:<4} | {res['tier']:<25} | Rs. {res['final_price']:>6.2f}")
print("=" * 65)
````

## 5. Leap Year Verification Algorithm

A year is a **Leap Year** if:
1. It is divisible by 4, **AND**
2. It is **NOT** divisible by 100, **UNLESS** it is also divisible by 400.

$$\text{IsLeap} = (\text{year} \% 4 == 0) \;\mathbf{and}\; (\text{year} \% 100 \ne 0 \;\mathbf{or}\; \text{year} \% 400 == 0)$$

````python
# Leap Year Verification Suite
def is_leap_year(year):
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        return True
    return False

test_years = [1900, 2000, 2020, 2024, 2025, 2100, 2400]
print("Leap Year Evaluation:")
for y in test_years:
    status = "LEAP YEAR" if is_leap_year(y) else "COMMON YEAR"
    print(f"  Year {y:<5}: {status}")
````

## 6. Structural Pattern Matching: `match..case` (Python 3.10+)

Introduced in Python 3.10, `match..case` provides powerful pattern matching that goes far beyond traditional `switch-case` statements in other languages:

### Key Capabilities:
- **Literal Matching**: Exact value equality.
- **Pipe Patterns (`|`)**: Matching any one of multiple alternatives (`case "add" | "+":`).
- **Wildcard Default (`case _:`):** Catches any value not matched by preceding patterns.
- **Guard Clauses (`if`):** Adds boolean conditions to patterns (`case [x, y] if x == y:`).
- **Sequence and Mapping Destructuring:** Automatically extracts values from lists, tuples, and dictionaries.

````python
# Structural Pattern Matching Master Suite (Match-Case)

# 1. Arithmetic Menu Dispatcher with Alternative Tokens
def calculate(op, a, b):
    match op.strip().lower():
        case "add" | "+" | "plus":
            return a + b
        case "sub" | "-" | "minus":
            return a - b
        case "mul" | "*" | "multiply":
            return a * b
        case "div" | "/" | "divide":
            return a / b if b != 0 else "Error: Division by Zero"
        case _:
            return f"Error: Unknown operator '{op}'"

print("--- Arithmetic Match-Case Dispatch ---")
print("10 + 5  :", calculate("+", 10, 5))
print("10 * 5  :", calculate("multiply", 10, 5))
print("10 / 0  :", calculate("/", 10, 0))

# 2. Shape Tuple Destructuring with Guard Conditions
def describe_shape(shape_tuple):
    match shape_tuple:
        case ("circle", r):
            return f"Circle with radius {r} (Area: {3.14159 * r**2:.2f})"
        case ("rectangle", w, h) if w == h:
            return f"Square with side length {w} (Area: {w * h})"
        case ("rectangle", w, h):
            return f"Rectangle with dimensions {w} x {h} (Area: {w * h})"
        case ("point",):
            return "Point with zero dimensions"
        case _:
            return "Unknown Shape Specification"

print("\n--- Destructuring & Guard Match-Case ---")
print(describe_shape(("circle", 5)))
print(describe_shape(("rectangle", 8, 8)))   # Triggered square guard!
print(describe_shape(("rectangle", 12, 6)))
print(describe_shape(("point",)))
````

## 7. The Guard Clause Pattern (Eliminating the Arrow Anti-Pattern)

### The Arrow Anti-Pattern (Deeply Nested Indentation):
When multiple preconditions are tested using nested `if` statements, code drifts further and further to the right, creating hard-to-maintain "arrow code":

```python
# The Arrow Anti-Pattern (AVOID):
def process_order(order):
    if order is not None:
        if order.is_valid():
            if order.has_stock():
                if order.is_paid():
                    ship_order(order)
```

### The Clean Solution: Guard Clauses (Early Exits):
Invert the conditions, handle invalid cases immediately with an early return or exception, and keep the main business logic cleanly at the base indentation level!

````python
# Guard Clause / Bouncer Pattern Implementation
def process_customer_checkout(user, cart, payment_token):
    # Guard 1: Validate User Authentication
    if not user.get("is_authenticated"):
        return {"status": "FAILED", "reason": "User not authenticated"}

    # Guard 2: Validate Cart Contents
    if not cart:
        return {"status": "FAILED", "reason": "Cart is empty"}

    # Guard 3: Validate Payment Token
    if not payment_token or len(payment_token) < 8:
        return {"status": "FAILED", "reason": "Invalid payment token"}

    # Happy Path (Executes cleanly at root level without deep nesting!)
    total_amount = sum(item["price"] * item["qty"] for item in cart)
    return {
        "status": "SUCCESS",
        "user": user["name"],
        "total": total_amount,
        "items_count": len(cart)
    }

# Testing the checkout pipeline
mock_user = {"name": "Pooja", "is_authenticated": True}
mock_cart = [{"name": "Book", "price": 450, "qty": 2}, {"name": "Pen", "price": 50, "qty": 3}]
result = process_customer_checkout(mock_user, mock_cart, "tok_secure_88899")
print("Checkout Pipeline Result:", result)
````

## 8. Dictionary Dispatch Table Pattern ($O(1)$ Branching)

When dealing with large sets of discrete operations, long `if-elif-else` ladders execute with $O(n)$ linear complexity.
The **Dictionary Dispatch Pattern** maps action keys directly to functions or lambdas, executing in **$O(1)$ constant time**.

````python
# Dictionary Dispatch Table Pattern
def action_deposit(state, amount):
    state["balance"] += amount
    return f"Deposited Rs. {amount}. New Balance: Rs. {state['balance']}"

def action_withdraw(state, amount):
    if state["balance"] >= amount:
        state["balance"] -= amount
        return f"Withdrew Rs. {amount}. New Balance: Rs. {state['balance']}"
    return "Insufficient Balance!"

def action_statement(state, _):
    return f"Account Statement: Balance = Rs. {state['balance']}"

# Dispatch Table Mapping: Key -> Function Pointer
dispatch_table = {
    "DEPOSIT": action_deposit,
    "WITHDRAW": action_withdraw,
    "STATEMENT": action_statement
}

account_state = {"balance": 10000.0}

def execute_banking_command(command, amount=0):
    handler = dispatch_table.get(command.upper())
    if handler:
        return handler(account_state, amount)
    return "Error: Unknown Command!"

print(execute_banking_command("DEPOSIT", 2500.0))
print(execute_banking_command("WITHDRAW", 4000.0))
print(execute_banking_command("STATEMENT"))
````

## 9. Advanced Structural Pattern Matching: Sequence Unpacking

Python 3.10+ `match..case` can unpack sequences with slicing stars (`*rest`), matching heads, tails, and prefixes.

````python
# Sequence Unpacking Pattern Matching
def parse_network_packet(packet):
    match packet:
        case ["HTTP", 200, *body]:
            return f"HTTP OK with {len(body)} body items: {body}"
        case ["HTTP", 404, *details]:
            return f"HTTP NOT FOUND: {details}"
        case ["LOG", level, source, message]:
            return f"[{level}] from {source}: {message}"
        case [protocol, *rest]:
            return f"Generic protocol '{protocol}' with payload: {rest}"
        case _:
            return "Malformed Packet"

print(parse_network_packet(["HTTP", 200, "json_payload", "headers", "cookies"]))
print(parse_network_packet(["LOG", "ERROR", "AuthService", "Connection timeout"]))
print(parse_network_packet(["FTP", "get", "file.txt"]))
````

## 10. Tricky Gotchas & Common Pitfalls

### ⚠️ Gotcha 1: Implicit Boolean Truthiness Traps
Beginners often check `if value == True:` instead of `if value:`.
In Python, `1 == True` evaluates to `True`, but `2 == True` evaluates to `False` even though `bool(2)` is truthy!

````python
# Truthiness vs Explicit Equality Gotcha
x = 2

# Trap:
print(f"bool({x}) is truthy?      -> {bool(x)}")
print(f"{x} == True?             -> {x == True} (Dangerous equality comparison!)")

# Best Practice:
if x:
    print(f"Correct: 'if x:' correctly treats {x} as truthy.")
````

### ⚠️ Gotcha 2: Mutually Exclusive `if-elif` vs. Sequential `if-if`
Replacing `elif` with a second independent `if` causes both blocks to execute if both conditions are met!

````python
# Chained if vs if-elif Trap
val = 85

print("--- Sequential 'if - if' (Double Execution Bug) ---")
if val >= 50:
    print("  Student Passed!")
if val >= 75:
    print("  Student Achieved Distinction!")  # Both execute!

print("\n--- Mutually Exclusive 'if - elif' (Correct Flow) ---")
if val >= 75:
    print("  Student Achieved Distinction!")
elif val >= 50:
    print("  Student Passed!")
````

## 11. Summary & Quick Revision Cheat Sheet

| Branching Construct | Syntax Signature | Core Purpose | Important Caveat |
|---|---|---|---|
| **Simple `if`** | `if condition:` | Conditional execution | If False, control skips block. |
| **`if..else`** | `if cond: ... else:` | Two-way binary branch | `else` has no condition. |
| **`if..elif..else`** | `if: ... elif: ... else:` | Multi-way cascading branch | Executes only the FIRST true branch. |
| **`match..case`** | `match val: case pattern:` | Pattern matching & destructuring | Python 3.10+ only. `case _:` is default. |
| **Ternary Operator** | `val_true if cond else val_false` | Inline conditional expression | Always requires both `if` and `else`. |
