## 1. Core Definitions & Program Philosophy

### What is a Program?
> **Definition**: A **Program** is a set of *optimized instructions* designed and executed to solve real-world problems.
>
> **What are Optimized Instructions?**
> Instructions are considered optimized when they satisfy two fundamental criteria:
> 1. **Minimal Memory Consumption**: Allocating only necessary space in RAM.
> 2. **Minimal Execution Time**: Completing computations with optimal CPU clock cycles ($O(1)$ or $O(n)$ algorithmic efficiency).

```
+--------------------------------------------------------------+
|                     OPTIMIZED PROGRAM                        |
|                                                              |
|   [ Less Execution Time ]   <--->   [ Less Memory Space ]    |
|                          \       /                           |
|                    Real-Time Problem                         |
|                         Solution                             |
+--------------------------------------------------------------+
```

## 3. The 11 Architectural Features of Python

Python provides **11 foundational features** that make it the industry standard for web services, data science, automation, and enterprise engineering:

1. **Simple & Readable**: Clean, pseudo-code-like syntax. Replaces complex punctuation with clear English keywords and mandatory indentation.
2. **Freeware & Open Source**: Zero licensing costs. Source code is freely accessible, inspectable, and modifiable under the PSF license.
3. **Platform Independent (WORA)**: Write Once, Run Anywhere. Python programs compile to portable bytecode executed by the platform-specific Python Virtual Machine (PVM).
4. **Dynamically Typed**: Variable types are determined automatically at runtime based on the assigned literal value. No explicit type declaration like `int x = 10;` is required.
5. **Interpreted Language**: Code is converted to bytecode and executed line-by-line by the PVM, enabling rapid prototyping and immediate feedback.
6. **High-Level Language**: Programmers focus on business logic without manually managing low-level hardware concerns such as memory allocation, pointer arithmetic, or CPU register layout.
7. **Robust (Strong)**: Built-in automated garbage collection (`gc`) prevents memory leaks, and comprehensive structured exception handling (`try-except`) catches runtime anomalies.
8. **Multi-Paradigm**: Fully supports both **Procedural / Functional Programming** (functions, lambdas) and **Object-Oriented Programming (OOPs)** (classes, inheritance, polymorphism).
9. **Extensible**: Native C/C++ code can be integrated seamlessly to speed up computationally heavy bottlenecks.
10. **Embedded**: Python scripts can be embedded directly inside C, C++, or Java host applications to provide scripting capabilities.
11. **Massive Ecosystem & Third-Party APIs**: Rich ecosystem of libraries including **NumPy, Pandas, SciPy, Scikit-learn, Matplotlib, PyTorch, TensorFlow, Django, and FastAPI**.

## 4. Execution Approaches: Interactive Mode vs. Batch Mode

In Python programming, code can be authored and executed using two distinct approaches:

### Approach 1: Interactive Mode
- Executed line-by-line via the interactive terminal shell (`python` CLI, IPython, or Jupyter Notebook).
- **Advantages**: Immediate evaluation, excellent for debugging, learning, and testing quick snippets.
- **Limitation**: Code is transient and not saved into permanent files for batch deployment.

### Approach 2: Batch Mode
- Source code is written into a text file with `.py` extension (e.g., `main.py`).
- Executed top-to-bottom via `python main.py`.
- **Advantages**: Permanent persistence, reusable modules, scheduled jobs, production deployment.

````python
# Demonstrating Batch Mode Execution Structure
def compute_salary(emp_id, emp_name, basic_pay, allowance_percent=20):
    """Simulates batch mode production payroll routine."""
    allowance = basic_pay * (allowance_percent / 100.0)
    gross_pay = basic_pay + allowance
    tax = gross_pay * 0.10  # 10% standard tax
    net_pay = gross_pay - tax
    
    return {
        "emp_id": emp_id,
        "name": emp_name,
        "basic": basic_pay,
        "allowance": allowance,
        "gross": gross_pay,
        "tax": tax,
        "net": net_pay
    }

record = compute_salary(101, "Aravind", 75000.0)
print("=" * 45)
print("            PAYROLL SUMMARY")
print("=" * 45)
for key, val in record.items():
    if isinstance(val, float):
        print(f"  {key.capitalize():<15}: Rs. {val:,.2f}")
    else:
        print(f"  {key.capitalize():<15}: {val}")
print("=" * 45)
````

## 5. CPython Architecture: Compilation vs. Interpretation

```
+------------------+       CPython Compiler       +---------------------+
|   Source Code    |  ------------------------->  |   Bytecode File     |
|   `script.py`    |                              |   `__pycache__/*.pyc|
+------------------+                              +---------------------+
                                                             |
                                                             | Read by PVM
                                                             v
+------------------+       Machine Code           +---------------------+
|  Target CPU / OS |  <-------------------------  |    Python Virtual   |
|  Hardware Runs   |       Executed Line-by-Line  |     Machine (PVM)   |
+------------------+                              +---------------------+
```

### The Two-Stage Process:
1. **Compilation Step**: When a `.py` file is executed or imported, CPython compiles human-readable code into intermediate **Bytecode** (`.pyc` stored in `__pycache__`).
2. **Interpretation Step**: The **Python Virtual Machine (PVM)** translates each bytecode instruction into native machine code instructions executed directly by the CPU.

````python
# Inspecting CPython Bytecode with 'dis' module
import dis

def calculate_discount(price, rate):
    discount = price * rate
    final_price = price - discount
    return final_price

print("Disassembled CPython Bytecode for calculate_discount:")
print("-" * 55)
dis.dis(calculate_discount)
print("-" * 55)
````

## 6. Everything is an Object: The CPython Object Model

In Python, numbers, strings, functions, and classes are first-class objects inherited from `object`.
Every object possesses:
1. **Identity (`id()`)**: Unique integer memory address of the C `PyObject` structure.
2. **Type (`type()`)**: Defines what operations can be performed on the object.
3. **Value**: The actual underlying state/data held in memory.
4. **Dynamic Attributes**: Custom user objects allow dynamic attribute assignment at runtime via `__dict__`.

````python
# Dynamic Attribute Binding and Object Inspection
class Student:
    pass

s = Student()  # Object creation

# Dynamically attaching attributes at runtime
s.sno = 100
s.sname = "Rossum"
s.marks = 92.5
s.college = "CWI Tech Institute"
s.city = "Amsterdam"

print("Object Identity (RAM Memory Address):", id(s))
print("Object Type Class                   :", type(s))
print("Instance Dynamic Attributes Dict    :", s.__dict__)
print(f"Formatted Output: Student {s.sname} (ID: {s.sno}) scored {s.marks}% at {s.college}")
````

## 9. Dependency Isolation: Python Virtual Environments (`venv`)

### Why Virtual Environments are Mandatory in Production:
By default, `pip install` installs packages globally into the system directory.
- **Problem**: Project A requires `Django 4.2`, while Project B requires `Django 5.1`. Global installation creates irreversible dependency conflicts and version pollution.
- **Solution**: A **Virtual Environment** creates a self-contained, isolated directory tree with its own Python executable and independent `site-packages` library folder.

### Core Lifecycle Commands:
```bash
# 1. Create a dedicated virtual environment named '.venv'
python -m venv .venv

# 2. Activate the virtual environment
# Windows PowerShell:
.venv\Scripts\Activate.ps1
# Windows Command Prompt:
.venv\Scripts\activate.bat
# Linux / macOS:
source .venv/bin/activate

# 3. Install dependencies inside the isolated environment
pip install requests pandas

# 4. Export exact pinned dependencies for reproducible builds
pip freeze > requirements.txt

# 5. Reinstall dependencies in a fresh environment
pip install -r requirements.txt

# 6. Deactivate when finished
deactivate
```

## 10. Tricky Gotchas & Common Pitfalls

### ⚠️ Gotcha 1: Integer Interning & Object Identity
CPython pre-allocates an array of small integer objects for values between `[-5, 256]`.
Integers in this range share the exact same memory address across the entire process!

````python
# Small Integer Caching Demonstration
a = 256
b = 256
print(f"a = 256, b = 256: a is b? -> {a is b} (Same ID: {id(a) == id(b)})")

x = 257
y = 257
print(f"x = 257, y = 257: x is y? -> {x is y} (IDs differ outside [-5, 256] range in separate statements)")
````

### ⚠️ Gotcha 2: Division Behavior in Python 2 vs. Python 3
In Python 3, `/` always performs **true float division**, while `//` performs **floor division**.

````python
print("True division  10 / 4  ->", 10 / 4)   # Always returns float: 2.5
print("Floor division 10 // 4 ->", 10 // 4)  # Truncates toward negative infinity: 2
print("Negative floor -10 // 4->", -10 // 4) # Rounds DOWN to -3, not -2!
````
