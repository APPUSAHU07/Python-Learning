```
Test Discovery
      │ Scans files matching test_*.py

Execution per Test:
   1. setUp()         -> Prepare clean test environment
   2. test_method()   -> Execute test & verify assertions
   3. tearDown()      -> Clean up resources (guaranteed)

      ▼
TestResult Collection & Reporting
```

## 🟢 3. Part 1: Beginner Fundamentals

### Step 1: Writing and Running Tests with `unittest`

````python
import unittest

def calculate_discount(price, discount_pct):
    if not (0 <= discount_pct <= 1):
        raise ValueError("Discount must be between 0 and 1")
    return round(price * (1 - discount_pct), 2)

class TestDiscount(unittest.TestCase):
    def test_standard_discount(self):
        self.assertEqual(calculate_discount(100.0, 0.20), 80.0)
        self.assertEqual(calculate_discount(50.0, 0.50), 25.0)
    
    def test_invalid_discount_raises_error(self):
        with self.assertRaises(ValueError):
            calculate_discount(100.0, 1.50)  # > 100%

# Running unittest suite inside Jupyter:
suite = unittest.TestLoader().loadTestsFromTestCase(TestDiscount)
runner = unittest.TextTestRunner(verbosity=2)
runner.run(suite)
````

## 🟡 4. Part 2: Intermediate Features

### Mocking External Dependencies with `unittest.mock`
Simulate network calls or databases without hitting real servers.

````python
from unittest.mock import Mock, patch

# Create mock object:
api_client = Mock()
api_client.fetch_user.return_value = {"id": 42, "name": "Alice"}

response = api_client.fetch_user(42)
print("Mocked response:", response)

# Assert that mock was called as expected:
api_client.fetch_user.assert_called_once_with(42)
print("Assertion passed: Mock was verified!")
````

## 🎓 Practical Hands-On & Real-World Examples

### 1. Banking Calculation & ICICI Module Unit Testing (`Test/icici.py`, `icicidemo.py`)
Building automated unit test suites using `unittest.TestCase` to validate interest rate algorithms and balance constraints.

````python
# Automated Unit Tests for Banking Calculations
import unittest

def calculate_simple_interest(principal, time_years, rate_percent):
    if principal < 0 or time_years < 0 or rate_percent < 0:
        raise ValueError("Principal, time, and rate must be non-negative!")
    interest = (principal * time_years * rate_percent) / 100.0
    total_amount = principal + interest
    return interest, total_amount

class TestBankingCalculations(unittest.TestCase):
    def test_standard_interest(self):
        interest, total = calculate_simple_interest(10000, 2, 7.5)
        self.assertEqual(interest, 1500.0)
        self.assertEqual(total, 11500.0)

    def test_zero_principal(self):
        interest, total = calculate_simple_interest(0, 5, 8.0)
        self.assertEqual(interest, 0.0)
        self.assertEqual(total, 0.0)

    def test_negative_values_raise_exception(self):
        with self.assertRaises(ValueError):
            calculate_simple_interest(-5000, 2, 7.5)

# Executing test runner programmatically in notebook
suite = unittest.TestLoader().loadTestsFromTestCase(TestBankingCalculations)
runner = unittest.TextTestRunner(verbosity=2)
result = runner.run(suite)
print(f"\nUnit Test Summary: Ran {result.testsRun} tests | Failures: {len(result.failures)} | Errors: {len(result.errors)}")
````
