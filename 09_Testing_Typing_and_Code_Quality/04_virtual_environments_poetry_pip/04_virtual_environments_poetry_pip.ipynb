## 🧠 2. CPython Architecture: How `venv` Isolates Packages

When a virtual environment is activated, it creates a `pyvenv.cfg` file pointing back to the base Python installation.
- When `python` starts, it looks for `pyvenv.cfg` next to the binary.
- It resets `sys.prefix` to the virtualenv directory.
- `sys.path` now prioritizes `.venv/lib/site-packages`, isolating your project from OS-wide libraries!

### Step 1: Standard `venv` Commands
```bash

# Activate on Linux/macOS:
source .venv/bin/activate

# Freeze dependencies:
python -m pip freeze > requirements.txt
```

## 🎓 Practical Hands-On & Real-World Examples

### 1. Environment & Package Metadata Introspection (`importlib.metadata`)
> **Core Principle**:
> - Modern Python standard library provides `importlib.metadata` to inspect installed distributions, package versions, and dependency requirements directly from Python code without shelling out to pip.

````python
# Programmatic Inspection of Environment Packages
import importlib.metadata as meta
import sys

print(f"Active Python Path: {sys.executable}")
print(f"In Virtual Environment: {sys.prefix != sys.base_prefix}\n")

# List first 8 installed distributions in active environment
installed = list(meta.distributions())[:8]
print("Sample of Installed Packages in Current Environment:")
for dist in installed:
    name = dist.metadata.get("Name", "Unknown")
    version = dist.metadata.get("Version", "Unknown")
    print(f"  - {name:<25} : Version {version}")
````

### 2. Dependency Specification & Version Constraint Validator
Simulating automated dependency verification against required version constraints.

````python
# Dependency Constraint Validator Simulation
required_specs = {
    "python": ">=3.10",
    "jupyterlab": ">=4.0",
    "requests": ">=2.28"
}

current_python = f"{sys.version_info.major}.{sys.version_info.minor}.{sys.version_info.micro}"
print(f"Environment Validation:")
print(f"  Current Python: {current_python} (Meets {required_specs['python']}) -> OK")
print("  Virtual Environment Prefix:", sys.prefix)
````
