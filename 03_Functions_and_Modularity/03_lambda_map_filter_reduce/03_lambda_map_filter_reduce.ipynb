## 1. Anonymous Functions (`lambda`): Syntax & Semantics

> **Definition**: A **Lambda Function** is a small, anonymous (unnamed) inline function defined using the `lambda` keyword:
>
> $$\text{Syntax}: \quad \mathbf{lambda} \; \text{param}_1, \text{param}_2, \dots : \text{single\_expression}$$
>
> **Core Constraints of Lambda Functions**:
> 1. Contains only a **single expression** (cannot contain multi-line statements, `return`, `while`, or `for`).
> 2. Automatically returns the evaluated result of the expression.
> 3. Intended for short-lived utility operations passed into higher-order functions.

```python
# Lambda Functions Practical Suite
# 1. Basic arithmetic lambda
add_op = lambda x, y: x + y
print("add_op(10, 20)        :", add_op(10, 20))

# 2. Lambda with Ternary conditional expression
max_finder = lambda a, b: a if a > b else b
print("max_finder(45, 92)    :", max_finder(45, 92))

# 3. Custom sorting key with lambda
inventory = [
    {"item": "Laptop", "price": 65000, "stock": 12},
    {"item": "Mouse", "price": 850, "stock": 45},
    {"item": "Monitor", "price": 18500, "stock": 8}
]

# Sort by price descending
sorted_by_price = sorted(inventory, key=lambda x: x["price"], reverse=True)
print("\nInventory Sorted by Price:")
for item in sorted_by_price:
    print(f"  {item['item']:<10}: Rs. {item['price']:>6,d} (Stock: {item['stock']})")
```

## 2. The `map()` Function: Element-Wise Transformation

> **Definition**: `map(function, iterable)` applies the specified transformation function to **every element** in the iterable and yields a lazy iterator stream of results ($O(1)$ memory consumption).

```python
# map() Function Demonstrations
# 1. Celsius to Fahrenheit temperature conversion
celsius_temps = [0.0, 20.0, 37.0, 100.0]
fahrenheit_stream = map(lambda c: (c * 9/5) + 32, celsius_temps)
print("Celsius Temps    :", celsius_temps)
print("Fahrenheit Temps :", list(fahrenheit_stream))

# 2. Multi-iterable map(): adding corresponding elements of two lists
list1 = [10, 20, 30, 40]
list2 = [1, 2, 3, 4]
summed_pairs = list(map(lambda a, b: a + b, list1, list2))
print("Summed Pairs     :", summed_pairs)
```

## 3. The `filter()` Function: Conditional Selection

> **Definition**: `filter(predicate_function, iterable)` evaluates each item against a boolean predicate function and yields only those items for which the predicate returns `True`.

```python
# filter() Function Demonstrations
raw_numbers = [12, -7, 0, 45, -23, 88, 101, -3, 64]

# 1. Filtering positive even numbers
pos_evens = list(filter(lambda x: x > 0 and x % 2 == 0, raw_numbers))
print("Original Numbers    :", raw_numbers)
print("Positive Even Filter:", pos_evens)

# 2. Filtering valid strings (removing empty, None, or whitespace strings)
messy_records = ["Alice", "", "Bob", None, "  ", "Charlie", "David"]
valid_records = list(filter(lambda s: s and s.strip(), messy_records))
print("Clean Valid Records :", valid_records)
```

## 4. The `reduce()` Function: Cumulative Aggregation

> **Definition**: In Python 3, `reduce` is located in the standard library `functools` module:
> `functools.reduce(function, iterable, [initializer])`
> It applies a function of two arguments cumulatively to the items of an iterable, reducing the sequence to a **single aggregate scalar value**.

```python
# functools.reduce() Demonstrations
from functools import reduce

numbers = [1, 2, 3, 4, 5, 6]

# 1. Cumulative multiplication (Factorial simulation)
factorial_result = reduce(lambda acc, val: acc * val, numbers)
print("Factorial Product via reduce:", factorial_result)

# 2. Finding maximum value in a list using reduce
max_value = reduce(lambda a, b: a if a > b else b, [45, 12, 89, 33, 97, 61])
print("Maximum Value via reduce    :", max_value)
```

## 5. End-to-End Functional Pipeline (`filter` $\rightarrow$ `map` $\rightarrow$ `reduce`)

Demonstrating the classic data engineering stream:
1. **Filter**: Select records meeting validation rules.
2. **Map**: Transform selected records.
3. **Reduce**: Aggregate into business metric.

```python
# Complete Functional Pipeline
from functools import reduce

transactions = [
    {"tx_id": 101, "amount": 4500.0, "status": "APPROVED", "tax_exempt": False},
    {"tx_id": 102, "amount": 12000.0, "status": "REJECTED", "tax_exempt": False},
    {"tx_id": 103, "amount": 8500.0, "status": "APPROVED", "tax_exempt": False},
    {"tx_id": 104, "amount": 1500.0, "status": "PENDING", "tax_exempt": False},
    {"tx_id": 105, "amount": 22000.0, "status": "APPROVED", "tax_exempt": True}
]

# Pipeline Step 1: Filter APPROVED transactions
approved_tx = filter(lambda tx: tx["status"] == "APPROVED", transactions)

# Pipeline Step 2: Map to calculate final amount (add 18% GST if not tax-exempt)
taxed_amounts = map(
    lambda tx: tx["amount"] * 1.18 if not tx["tax_exempt"] else tx["amount"],
    approved_tx
)

# Pipeline Step 3: Reduce to calculate total gross revenue
total_revenue = reduce(lambda acc, amt: acc + amt, taxed_amounts, 0.0)

print("=" * 45)
print("       FUNCTIONAL PIPELINE METRICS")
print("=" * 45)
print(f"  Total Processed Revenue: Rs. {total_revenue:,.2f}")
print("=" * 45)
```

## 6. Summary & Quick Revision Cheat Sheet

| Functional Tool | Module | Input $\rightarrow$ Output | Memory Strategy | Key Operational Role |
|---|---|---|---|---|
| `lambda` | Built-in keyword | Params $\rightarrow$ Single expression | Inline | Quick anonymous function. |
| `map()` | Built-in | $N$ items $\rightarrow$ $N$ transformed items | **Lazy Iterator ($O(1)$ RAM)** | Element-wise mapping. |
| `filter()` | Built-in | $N$ items $\rightarrow$ $\le N$ filtered items | **Lazy Iterator ($O(1)$ RAM)** | Predicate selection. |
| `reduce()` | `functools` | $N$ items $\rightarrow$ **1 scalar result** | Eager accumulation | Aggregations (sum, product, min/max). |
