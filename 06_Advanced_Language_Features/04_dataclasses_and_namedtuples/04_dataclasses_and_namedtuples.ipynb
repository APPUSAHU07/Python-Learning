## 📌 1. Topic Overview
Python provides clean, idiomatic tools for data-centric classes: `namedtuple` (immutable, tuple-based) and `dataclass` (feature-rich, mutable or immutable, auto-generating `__init__`, `__repr__`, `__eq__`).

---
## 🧠 2. CPython Architecture: `namedtuple` vs `@dataclass`

| Feature | `collections.namedtuple` | `@dataclasses.dataclass` |
| :--- | :--- | :--- |
| **Underlying Base** | Subclass of `tuple` | Standard Python class (`object`) |
| **Mutability** | **Immutable** | **Mutable** by default (`frozen=True` for immutable) |
| **Memory Footprint** | Extremely compact (same as tuple, no `__dict__`) | Standard `__dict__` (unless `slots=True`) |
| **Indexing & Unpacking** | Supports `pt[0]` and `x, y = pt` | No (attribute access only `pt.x`) |
| **Default Values & Types** | Basic | Advanced (`field`, `default_factory`, type hints) |

### How `@dataclass` Works
The `@dataclass` decorator inspects your class annotations (`__annotations__`) and dynamically generates CPython dunder methods (`__init__`, `__repr__`, `__eq__`, `__hash__`) by synthesizing Python source code and executing it via `exec()`.

## 🟢 3. Part 1: Beginner Fundamentals

### Step 1: `collections.namedtuple`
Lightweight, immutable, tuple-compatible records.

````python
from collections import namedtuple

Point = namedtuple("Point", ["x", "y"])
p = Point(10, 20)

print("Point object  :", p)
print("Named access  :", p.x, p.y)
print("Indexed access:", p[0], p[1])

# Unpacking:
x_coord, y_coord = p
print(f"Unpacked: {x_coord=}, {y_coord=}")
````

### Step 2: Modern `@dataclass` (Python 3.7+)
Auto-generates boilerplate constructors and comparison methods.

````python
from dataclasses import dataclass

@dataclass
class Product:
    name: str
    price: float
    quantity: int = 1
    
    def total_cost(self):
        return self.price * self.quantity

item1 = Product("Mechanical Keyboard", 120.0, 2)
item2 = Product("Mechanical Keyboard", 120.0, 2)

print("Auto __repr__:", item1)
print("Auto __eq__  :", item1 == item2)  # True!
print("Total cost   :", item1.total_cost())
````

## 🟡 4. Part 2: Intermediate Features

### `field(default_factory=...)` for Mutable Defaults
Dataclasses explicitly prevent mutable defaults (`tags: list = []`). You MUST use `default_factory`!

````python
from dataclasses import dataclass, field

@dataclass
class UserProfile:
    username: str
    # Correct mutable default:
    tags: list[str] = field(default_factory=list)

u1 = UserProfile("alice")
u2 = UserProfile("bob")
u1.tags.append("admin")

print("u1 tags:", u1.tags)
print("u2 tags:", u2.tags, "(Properly isolated!)")
````

### Immutable & Hashable Dataclasses (`frozen=True`)

````python
@dataclass(frozen=True)
class ServerConfig:
    ip: str
    port: int

cfg = ServerConfig("10.0.0.1", 443)
print("Hashable:", hash(cfg))

try:
    cfg.port = 8080
except Exception as e:
    print("FrozenInstanceError caught:", type(e).__name__)
````

## 📋 5. Part 3: Quick Revision Cheat Sheet

```
+------------------------------------+---------------------------------------+
| Construct                          | Key Rule                              |
+------------------------------------+---------------------------------------+
| namedtuple                         | Immutable, tuple-based, minimal memory|
| @dataclass                         | Generates __init__, __repr__, __eq__  |
| field(default_factory=list)        | Required for mutable default values   |
| @dataclass(frozen=True)            | Makes instances read-only & hashable  |
```

## 🎓 Practical Hands-On & Real-World Examples

### 1. Clean Data Models: `namedtuple` vs `@dataclass`
Eliminating boilerplate constructors, string representations, and equality checks for domain models.

````python
# Clean Data Containers: namedtuple vs dataclass
from collections import namedtuple
from dataclasses import dataclass

# 1. namedtuple: Lightweight, memory-efficient immutable container
StudentTuple = namedtuple("StudentTuple", ["sno", "sname", "course", "marks"])
st1 = StudentTuple(101, "Rajesh", "Python", 88.0)
print("NamedTuple Instance:", st1)
print(f"Access by attribute: {st1.sname}, Access by index: {st1[1]}")

# 2. dataclass: Modern mutable or frozen container with type annotations
@dataclass
class CourseRegistration:
    student_id: int
    student_name: str
    course: str
    fee_paid: float
    is_active: bool = True

    def calculate_tax(self, rate=0.18):
        return self.fee_paid * rate

reg1 = CourseRegistration(201, "Shalini", "Data Science", 15000.0)
print("\nDataClass Instance:", reg1)
print(f"Tax for registration: Rs. {reg1.calculate_tax():,.2f}")
````
