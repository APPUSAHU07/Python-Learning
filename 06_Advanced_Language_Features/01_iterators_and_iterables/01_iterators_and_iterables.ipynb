## 📌 1. Topic Overview
Iteration is central to Python's design. An **Iterable** is anything you can loop over. An **Iterator** is the stateful cursor that produces items sequentially on demand.

---
## 🧠 2. CPython Architecture: The Iteration Protocol Contract

```
┌────────────────────────────────────────────────────────┐
| Iterable: Has __iter__() returning an Iterator         |
|           (e.g., list, tuple, dict, set, range)        |
├────────────────────────────────────────────────────────┤
| Iterator: Has __next__() returning next value or       |
|           raising StopIteration. Also has __iter__()   |
|           returning self.                              |
└────────────────────────────────────────────────────────┘
```

### Fallback to Sequence Protocol (`__getitem__`)
If an object lacks `__iter__()`, Python checks for `__getitem__()` starting at index 0. If it finds `__getitem__`, it constructs a sequence iterator automatically!

## 🟢 3. Part 1: Beginner Fundamentals

### Step 1: `iter()` and `next()`
Converting an iterable into an iterator and consuming values manually.

````python
fruits = ["apple", "banana", "cherry"]

# 1. Get the iterator:
fruit_iterator = iter(fruits)
print("Iterator object:", fruit_iterator)

# 2. Step through with next():
print("Item 1:", next(fruit_iterator))
print("Item 2:", next(fruit_iterator))
print("Item 3:", next(fruit_iterator))

# 3. Calling next() again raises StopIteration:
try:
    next(fruit_iterator)
except StopIteration:
    print("StopIteration caught! All elements consumed.")
````

### Step 2: Building a Custom Iterator Class

````python
class Countdown:
    def __init__(self, start):
        self.current = start
    
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.current <= 0:
            raise StopIteration
        val = self.current
        self.current -= 1
        return val

for num in Countdown(4):
    print("Countdown:", num)
````

## 🟡 4. Part 2: Intermediate Features

### Two-Argument `iter(callable, sentinel)`
Repeatedly invokes `callable()` until it returns the `sentinel` value! Perfect for reading fixed-size chunks or sockets.

````python
import random

# Call random.randint(1, 6) until it rolls a 6:
rolls = list(iter(lambda: random.randint(1, 6), 6))
print("Rolled numbers before rolling 6:", rolls)
````

## 🔴 5. Part 3: Tricky Parts, Gotchas & Interview Traps

### ⚠️ Gotcha 1: Iterators are One-Shot Containers
An iterator exhausts itself after a single loop. An iterable (like a list) can be looped over infinitely many times!

````python
it = iter([1, 2, 3])
print("Loop 1:", [x for x in it])
print("Loop 2:", [x for x in it], "<- Empty! Iterator is exhausted.")
````

## 🎓 Practical Hands-On & Real-World Examples

### 1. Converting Iterables into Iterators & `StopIteration` (`IterEx1.py` - `IterEx3.py`)
> **Core Iterator Principles**:
> - An **Iterable** is any object that can be looped over (e.g. `tuple`, `set`, `list`, `dict`, `str`).
> - An **Iterator** is a stateful stream produced by `iter(iterable_obj)`.
> - Elements are fetched one by one via `next(iterator_obj)`.
> - When elements are exhausted, Python raises `StopIteration`.

````python
# Practical Implementation: IterEx2.py & IterEx3.py
# Converting tuple & set to iterator objects
sample_tuple = (10, "Rossum", 45.67, True, 2 + 3j)
tuple_iter = iter(sample_tuple)

print("Original Object Type: ", type(sample_tuple))
print("Iterator Object Type: ", type(tuple_iter))

print("\nIterating sequentially using next():")
while True:
    try:
        val = next(tuple_iter)
        print(f"  Fetched item: {val} (Type: {type(val).__name__})")
    except StopIteration:
        print("Reached End of Stream: StopIteration gracefully caught.")
        break
````

### 2. Custom Countdown Iterator Class
Implementing the `__iter__()` and `__next__()` protocol from scratch.

````python
# Custom Iterator Protocol Implementation
class CountdownIterator:
    def __init__(self, start_val):
        self.current = start_val

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= 0:
            raise StopIteration
        val = self.current
        self.current -= 1
        return val

cd = CountdownIterator(5)
print("Countdown Sequence:", list(cd))
````
