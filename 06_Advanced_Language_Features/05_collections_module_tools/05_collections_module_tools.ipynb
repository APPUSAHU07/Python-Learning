## 🧠 2. CPython Architecture: `list` vs `collections.deque`

### Why You Should NEVER Use `list` as a Queue
- **`list.pop(0)` or `list.insert(0, x)`**: Takes **$O(n)$ linear time**! CPython must shift every single pointer in the memory block.
- **`collections.deque`**: Implemented as a **doubly-linked list of 64-element block arrays**. Inserting or popping from either end (`appendleft()`, `popleft()`) is guaranteed **$O(1)$ constant time**!

```
            deque Doubly-Linked Block Architecture
┌────────────────┐       ┌────────────────┐       ┌────────────────┐
| Block (64 ptrs)| <---> | Block (64 ptrs)| <---> | Block (64 ptrs)|
└────────────────┘       └────────────────┘       └────────────────┘
  ^                                                 ^
  popleft() / appendleft() [O(1)]                   pop() / append() [O(1)]
```

## 🟢 3. Part 1: Beginner Fundamentals

### Step 1: `collections.deque` (High-Speed Queues & Sliding Windows)

````python
from collections import deque

# O(1) double-ended queue:
q = deque(["task2", "task3"])
q.appendleft("task1")  # O(1) insert at head
q.append("task4")      # O(1) insert at tail
print("Deque contents:", q)

first = q.popleft()    # O(1) remove from head
print("Popped head   :", first)

# Sliding window / capped queue with maxlen:
recent_logs = deque(maxlen=3)
for i in range(1, 6):
    recent_logs.append(f"Log {i}")
    print("Current logs (maxlen=3):", list(recent_logs))
````

### Step 2: `collections.Counter` (Frequency Counting & Multiset Math)

````python
from collections import Counter

words = ["apple", "banana", "apple", "orange", "banana", "apple"]
counter = Counter(words)

print("Counts            :", counter)
print("Most common 2 items:", counter.most_common(2))

# Counter arithmetic:
c1 = Counter(a=3, b=1)
c2 = Counter(a=1, b=2)
print("c1 + c2:", c1 + c2)
print("c1 - c2:", c1 - c2)
````

## 🟡 4. Part 2: Intermediate Features

### `collections.ChainMap`: Grouping Multiple Mappings
Searches across multiple dictionaries without merging them (ideal for config overrides: CLI args -> Env vars -> Defaults).

````python
from collections import ChainMap

defaults = {"theme": "light", "debug": False, "port": 8080}
env_vars = {"debug": True}
cli_args = {"port": 9000}

# ChainMap searches in priority order (left to right):
config = ChainMap(cli_args, env_vars, defaults)
print("Resolved port  :", config["port"])   # 9000 from CLI
print("Resolved debug :", config["debug"])  # True from Env
print("Resolved theme :", config["theme"])  # light from Defaults
````

## 🎓 Practical Hands-On & Real-World Examples

### 1. High-Performance Container Tools (`Counter`, `defaultdict`, `deque`)
Solving frequency counting, grouping, and double-ended queue operations cleanly.

````python
# Collections Module Drills
from collections import Counter, defaultdict, deque

# 1. Counter: Word and character frequency analysis
text = "python programming in data science and python web development"
word_counts = Counter(text.split())
print("Top 3 Most Common Words:", word_counts.most_common(3))

# 2. defaultdict: Grouping without key checking
students = [("CS", "Amit"), ("ECE", "Bina"), ("CS", "Charan"), ("MECH", "David"), ("ECE", "Divya")]
dept_groups = defaultdict(list)
for dept, name in students:
    dept_groups[dept].append(name)

print("\nGrouped Students by Department:")
for dept, names in dept_groups.items():
    print(f"  {dept:<6}: {names}")

# 3. deque: High-speed O(1) appends and pops on both ends
task_queue = deque(["Task-1", "Task-2", "Task-3"])
task_queue.appendleft("VIP-Task-0")
task_queue.append("Task-4")
print(f"\nDeque Queue: {task_queue}")
print(f"Popped from left: {task_queue.popleft()}")
````
