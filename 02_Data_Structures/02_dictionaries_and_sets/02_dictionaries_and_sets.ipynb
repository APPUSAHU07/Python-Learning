## 1. Core Purpose & Definitions

### What is a Set?
> A **Set** is an **unordered**, **mutable collection** of **unique**, **hashable elements** enclosed in curly braces `{ }`.
> - **Unordered**: Elements have no positional index; items cannot be accessed via `s[0]`.
> - **Unique**: Automatically eliminates duplicate entries upon insertion.
> - **Hashable Elements Only**: Every element in a set must be immutable (e.g. `int`, `str`, `tuple`). A set *cannot* contain lists or dictionaries.

### What is a FrozenSet?
> A **FrozenSet** is an **immutable set**.
> - Once created, it cannot be modified (no `.add()`, `.remove()`, or `.clear()`).
> - Because it is immutable, a `frozenset` **is hashable**!
> - Crucial Capability: A `frozenset` can be **nested inside another set** or used as a **dictionary key**!

### What is a Dictionary (`dict`)?
> A **Dictionary** is a **mutable**, **dynamic mapping** of unique, hashable **keys** to arbitrary **values** (`{key: value}`).
> - **Insertion-Ordered**: Since Python 3.7+, dictionaries strictly preserve the order in which keys were inserted.
> - **$O(1)$ Hash Table Lookup**: Average constant time complexity for key lookup, insertion, and deletion.

## 2. Set Theory Operations & Predefined Methods

Python sets natively implement all mathematical set theory operations through both operators and method names:

| Set Operation | Operator | Equivalent Method | Mathematical Meaning |
|---|---|---|---|
| **Union** | `A \| B` | `A.union(B)` | All elements present in `A` or `B` (or both). |
| **Intersection** | `A & B` | `A.intersection(B)` | Elements present in **both** `A` and `B`. |
| **Difference** | `A - B` | `A.difference(B)` | Elements in `A` that are **not** in `B`. |
| **Symmetric Diff**| `A ^ B` | `A.symmetric_difference(B)` | Elements in `A` or `B`, but **not in both**. |
| **Subset Test** | `A <= B` | `A.issubset(B)` | `True` if all elements of `A` exist in `B`. |
| **Superset Test**| `A >= B` | `A.issuperset(B)` | `True` if `A` contains all elements of `B`. |
| **Disjoint Test**| None | `A.isdisjoint(B)` | `True` if `A` and `B` have zero elements in common. |

```python
# Set Theory Operations Execution Suite
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

print(f"Set A: {set_a}")
print(f"Set B: {set_b}")
print("-" * 50)
print(f"Union (A | B)               : {set_a | set_b}")
print(f"Intersection (A & B)        : {set_a & set_b}")
print(f"Difference (A - B)          : {set_a - set_b}")
print(f"Difference (B - A)          : {set_b - set_a}")
print(f"Symmetric Difference (A ^ B): {set_a ^ set_b}")

# Subset and Disjoint checks
subset_test = {1, 2}
print(f"\n{subset_test} issubset of Set A?  : {subset_test.issubset(set_a)}")
print(f"Set A isdisjoint with {9, 10}? : {set_a.isdisjoint({9, 10})}")
```

## 3. Set Mutation Methods & Frozenset Architecture

### Mutation Methods:
- `.add(elem)`: Inserts a single item.
- `.remove(elem)`: Removes `elem`; **raises `KeyError` if missing**.
- `.discard(elem)`: Removes `elem` safely; **does NOT raise error** if missing!
- `.pop()`: Removes and returns an arbitrary element.
- `.clear()`: Removes all items, leaving `set()`.

### Frozenset Nesting Rule:
You **cannot** place a `set` inside a `set` (raises `TypeError: unhashable type: 'set'`).
However, you **CAN** place a `frozenset` inside a `set`!

```python
# Set Mutation & Frozenset Nesting
# 1. remove vs discard safe removal
tech_stack = {"Python", "SQL", "Docker", "Git"}
tech_stack.add("Kubernetes")
print("After add():", tech_stack)

# discard() never crashes if item is missing
tech_stack.discard("NonExistentTech")  # Safe!
print("After safe discard():", tech_stack)

# 2. Frozenset: Immutable & Hashable
immutable_group = frozenset(["Admin", "Editor", "Viewer"])
print(f"\nFrozenset: {immutable_group} (Type: {type(immutable_group).__name__})")

# Valid Nested Set using frozensets!
nested_roles_set = {
    frozenset({"Read", "Write"}),
    frozenset({"Read", "Execute"}),
    frozenset({"FullControl"})
}
print("Set of nested frozensets initialized successfully:", nested_roles_set)
```

## 4. Dictionary Architecture & The Complete Methods Suite

### CPython Compact Hash Table Architecture:
Since Python 3.6/3.7, CPython uses a two-table design:
1. **Sparse Array of Indices**: Stores bucket indices based on hash codes.
2. **Dense Array of Entries**: Stores `[hash, key, value]` sequentially in insertion order.
This innovation reduced dictionary RAM usage by 30%–40% and made dictionaries ordered by default!

| Method | Operation Performed | Return Value | Notes |
|---|---|---|---|
| `.get(key, [default])` | Safe value access | Value or `default` (`None`) | **Never raises KeyError**. |
| `.setdefault(key, default)` | Returns value if key exists; else inserts `key: default` | Current or new value | Great for grouping without importing collections. |
| `.pop(key, [default])` | Removes key and returns its value | Value or `default` | Raises `KeyError` if key missing and no default. |
| `.popitem()` | Removes and returns last inserted `(key, value)` | Tuple `(k, v)` | LIFO order. |
| `.update(other_dict)` | Merges keys/values from another dict | `None` | Overwrites matching keys in-place. |
| `.keys()`, `.values()`, `.items()` | Returns dynamic view objects | View object | Automatically reflects subsequent dictionary mutations. |

```python
# Complete Dictionary Operations Suite
student = {
    "roll": 101,
    "name": "Kalyan",
    "course": "Full Stack Python",
    "city": "Hyderabad"
}

# 1. Safe access via get()
phone = student.get("phone", "Not Provided")
print("Safe get() phone:", phone)

# 2. setdefault(): Returns existing or sets default
student.setdefault("city", "Bangalore")    # 'city' already exists -> remains 'Hyderabad'
student.setdefault("grade", "A+")          # 'grade' does not exist -> inserted!
print("After setdefault calls:", student)

# 3. Dynamic dictionary views
keys_view = student.keys()
items_view = student.items()
print("\nKeys view :", list(keys_view))

# 4. Dictionary Merging (Python 3.9+ Merge Operator |)
additional_info = {"email": "kalyan@tech.org", "batch": "2026-Q1"}
merged_student = student | additional_info  # Creates brand new merged dictionary!
print("\nMerged Dictionary (Python 3.9+ '|' operator):")
for k, v in merged_student.items():
    print(f"  {k:<12}: {v}")
```

## 5. Practical Algorithms: Word Counter & Inverting Dictionaries

### Algorithm 1: Character & Word Frequency Analysis
Constructing frequency distributions using standard dictionary lookup and update idioms.

### Algorithm 2: Inverting a Dictionary (Swapping Keys and Values)
Mapping values back to keys, with collision handling using lists.

```python
# Dictionary Frequency & Inversion Algorithms
# 1. Character Frequency Counter
phrase = "data science and data engineering in python"
freq_map = {}
for char in phrase:
    if char != " ":
        freq_map[char] = freq_map.get(char, 0) + 1

print("Top Character Frequencies:")
for ch, count in sorted(freq_map.items(), key=lambda item: item[1], reverse=True)[:5]:
    print(f"  '{ch}': {count} occurrences")

# 2. Inverting a Dictionary with duplicate values handled
original_roles = {
    "Alice": "Engineering",
    "Bob": "Design",
    "Charlie": "Engineering",
    "David": "Product",
    "Eve": "Design"
}

dept_to_members = {}
for member, dept in original_roles.items():
    dept_to_members.setdefault(dept, []).append(member)

print("\nInverted Dictionary (Departments -> Member Lists):")
for dept, members in dept_to_members.items():
    print(f"  {dept:<15}: {members}")
```

## 6. Tricky Gotchas & Common Pitfalls

### ⚠️ Gotcha 1: The Empty Set `{}` Trap
Writing `s = {}` creates an **empty dictionary**, NOT an empty set!
To create an empty set, you **MUST** use the constructor function: `s = set()`.

```python
# The Empty Set Trap
empty_curly = {}
empty_func = set()

print(f"empty_curly = {{}}  -> Type: {type(empty_curly).__name__} (It's a Dictionary!)")
print(f"empty_func = set() -> Type: {type(empty_func).__name__} (Actual Empty Set)")
```

### ⚠️ Gotcha 2: Mutating Dictionary Size During Iteration
Modifying keys (adding or deleting items) while iterating over a dictionary directly raises a `RuntimeError`!

```python
# Dictionary Mutation During Iteration Trap
data = {"a": 1, "b": 2, "c": 3, "d": 4}

# TRAP (Bug!):
try:
    for k in data:
        if data[k] % 2 == 0:
            del data[k]  # Modifying during loop!
except RuntimeError as err:
    print(f"Caught Expected Mutation Error: {err}")

# CORRECT PRODUCTION FIX: Iterate over a snapshot list of keys!
data_safe = {"a": 1, "b": 2, "c": 3, "d": 4}
for k in list(data_safe.keys()):  # Creates a static list copy
    if data_safe[k] % 2 == 0:
        del data_safe[k]
print("Safely filtered dictionary:", data_safe)
```
