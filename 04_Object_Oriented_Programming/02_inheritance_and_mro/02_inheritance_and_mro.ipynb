## 1. Core Purpose & The 5 Types of Inheritance

> **Definition**: **Inheritance** is the object-oriented mechanism by which a Derived (Child / Sub) Class inherits all accessible data members and methods from a Base (Parent / Super) Class.
>
> **The 5 Classical Types of Inheritance**:
> 1. **Single Inheritance**: One child inherits from one parent.
> 2. **Multi-Level Inheritance**: Chain of inheritance ($A \rightarrow B \rightarrow C$).
> 3. **Hierarchical Inheritance**: Multiple children inherit from a single parent.
> 4. **Multiple Inheritance**: One child inherits from multiple parent classes.
> 5. **Hybrid Inheritance**: Combination of multiple inheritance patterns forming a diamond graph.

## 2. Multi-Level Academic Hierarchy: Constructor Chaining with `super()`

In deep multi-level hierarchies, each derived class must invoke `super().__init__()` to forward parameters up the inheritance chain, ensuring that every ancestor layer initializes its own state safely.

```
+------------------------+
|      University        |  (u_name, u_location)
+------------------------+
            |
            v
+------------------------+
|       College          |  (c_name, c_code)
+------------------------+
            |
            v
+------------------------+
|       Student          |  (s_roll, s_name)
+------------------------+
```

````python
# Multi-Level Inheritance & Constructor Chaining Suite
class University:
    def __init__(self, u_name, u_location):
        self.u_name = u_name
        self.u_location = u_location

    def get_univ_info(self):
        return f"University: {self.u_name} ({self.u_location})"

class College(University):
    def __init__(self, u_name, u_location, c_name, c_code):
        super().__init__(u_name, u_location)  # Forwards up to University
        self.c_name = c_name
        self.c_code = c_code

    def get_college_info(self):
        return f"{self.get_univ_info()} | College: {self.c_name} [{self.c_code}]"

class Student(College):
    def __init__(self, u_name, u_location, c_name, c_code, s_roll, s_name):
        super().__init__(u_name, u_location, c_name, c_code)  # Forwards up to College
        self.s_roll = s_roll
        self.s_name = s_name

    def display_profile(self):
        print("=" * 65)
        print("               ACADEMIC PROFILE REPORT")
        print("=" * 65)
        print("Institutional Chain :", self.get_college_info())
        print(f"Student Details     : Roll #{self.s_roll} | Name: {self.s_name}")
        print("=" * 65)

stu = Student("Osmania University", "Hyderabad", "CBIT Engineering", "CBIT-01", 101, "Vikram Sharma")
stu.display_profile()
````

## 3. Multiple Inheritance, The Diamond Problem & C3 Linearization (MRO)

### What is the Diamond Problem?
When class `D` inherits from both `B` and `C`, and both `B` and `C` inherit from class `A`:
If both `B` and `C` override a method from `A`, which version does `D` execute?

### C3 Linearization (Method Resolution Order):
Python resolves method calls using the **C3 Linearization Algorithm**:
- Inspect the class itself first.
- Inspect parent classes left-to-right as listed in the class definition.
- Ensure ancestors appear only after their children.
- Inspectable at runtime via `Class.__mro__` or `Class.mro()`.

````python
# The Diamond Problem & MRO Inspection Suite
class Device:
    def operate(self):
        return "Device: Base hardware operation"

class Scanner(Device):
    def operate(self):
        return "Scanner: Optical digitization scanning"

class Printer(Device):
    def operate(self):
        return "Printer: Thermal ink printing"

class MultiFunctionCopier(Scanner, Printer):
    pass

copier = MultiFunctionCopier()
print("copier.operate() Output :", copier.operate())  # Scanner wins because Scanner appears before Printer!

print("\nMethod Resolution Order (MRO) for MultiFunctionCopier:")
for idx, cls in enumerate(MultiFunctionCopier.__mro__):
    print(f"  Step {idx}: {cls.__name__}")
````

| **Inheritance Syntax** | `class Child(Parent):` | Reuses state & behavior. |
