## 1. Dunder Protocol Philosophy & Core Concepts

> **Definition**: **Magic Methods** (colloquially called **Dunder Methods** for *Double Underscore*) are predefined special methods in Python that begin and end with two underscores (`__method__`).
>
> **Core Characteristic**:
> You rarely invoke magic methods directly. Instead, the **Python interpreter automatically invokes them behind the scenes** when specific syntactic operations, operators, or built-in functions are applied to an object.
>
> | Built-in Expression | Implicit Python Dunder Hook |
> |---|---|
> | `str(obj)` or `print(obj)` | `obj.__str__()` |
> | `repr(obj)` | `obj.__repr__()` |
> | `a + b` | `a.__add__(b)` |
> | `a == b` | `a.__eq__(b)` |
> | `len(obj)` | `obj.__len__()` |
> | `obj[key]` | `obj.__getitem__(key)` |
> | `obj()` (Calling object like a function) | `obj.__call__()` |

## 2. Object Lifecycle: `__new__` vs. `__init__` vs. `__del__`

1. **`__new__(cls, ...)`**: The **true creator/allocator**. Static method that returns a brand new empty heap object. (Used in Metaclasses and Singletons).
2. **`__init__(self, ...)`**: The **initializer**. Receives the newly allocated instance and binds attributes.
3. **`__del__(self)`**: The **destructor**. Invoked automatically by CPython when the object's reference count drops to zero.

```python
# Object Lifecycle Demonstration (__new__, __init__, __del__)
class LifecycleDemo:
    def __new__(cls, *args, **kwargs):
        print("1. [__new__] Memory space allocated on heap for new instance.")
        instance = super().__new__(cls)
        return instance

    def __init__(self, label):
        self.label = label
        print(f"2. [__init__] Instance state initialized for '{self.label}'.")

    def __del__(self):
        print(f"3. [__del__] Destructor called. Releasing resource '{self.label}' from RAM.")

obj = LifecycleDemo("Worker-Alpha")
del obj  # Decrements ref count to 0 -> triggers __del__ immediately!
```

## 3. String Representation: `__str__` vs. `__repr__`

- **`__str__(self)`**: Intended for **end-users**. Should be readable, intuitive, and clean.
- **`__repr__(self)`**: Intended for **developers and debugging**. Should be unambiguous, and ideally match valid Python code that could recreate the object: `eval(repr(obj)) == obj`.

## 4. Operator Overloading & Container Protocol Suite

Implementing custom 2D Vectors with arithmetic (`+`, `-`), comparisons (`==`), length (`abs`), indexing (`[]`), and callable functionality (`()`).

```python
# Complete Vector2D Operator Overloading Suite
class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # User-facing string representation
    def __str__(self):
        return f"({self.x}, {self.y})"

    # Developer-facing debug representation
    def __repr__(self):
        return f"Vector2D(x={self.x}, y={self.y})"

    # Operator + (Addition)
    def __add__(self, other):
        if not isinstance(other, Vector2D):
            return NotImplemented
        return Vector2D(self.x + other.x, self.y + other.y)

    # Operator - (Subtraction)
    def __sub__(self, other):
        if not isinstance(other, Vector2D):
            return NotImplemented
        return Vector2D(self.x - other.x, self.y - other.y)

    # Operator == (Equality)
    def __eq__(self, other):
        if not isinstance(other, Vector2D):
            return False
        return self.x == other.x and self.y == other.y

    # Container indexing: v[0] -> x, v[1] -> y
    def __getitem__(self, index):
        if index == 0:
            return self.x
        elif index == 1:
            return self.y
        raise IndexError("Vector2D index out of range (use 0 or 1).")

    # Callable instance: v(scale_factor)
    def __call__(self, scale):
        return Vector2D(self.x * scale, self.y * scale)

v1 = Vector2D(3, 4)
v2 = Vector2D(1, 2)

print(f"v1               : {v1}")
print(f"v2               : {v2}")
print(f"v1 + v2 (__add__): {v1 + v2}")
print(f"v1 - v2 (__sub__): {v1 - v2}")
print(f"v1 == v2 (__eq__): {v1 == v2}")
print(f"v1[0] (Indexing) : {v1[0]}")
print(f"v1(3) (__call__) : {v1(3)} (Scaled by 3)")
print(f"repr(v1)         : {repr(v1)}")
```

## 5. Summary & Quick Revision Cheat Sheet

| Category | Dunder Method | Hook Expression | Role |
|---|---|---|---|
| **Allocation** | `__new__(cls)` | `Class()` | Allocates memory for object. |
| **Initialization** | `__init__(self)` | `Class()` | Binds initial attribute values. |
| **Cleanup** | `__del__(self)` | `del obj` (when ref count=0) | Releases file/DB handles. |
| **Human String** | `__str__(self)` | `str(obj)`, `print(obj)` | User-readable string. |
| **Machine String** | `__repr__(self)` | `repr(obj)`, REPL display | Exact code representation. |
| **Arithmetic** | `__add__`, `__sub__`, `__mul__` | `+`, `-`, `*` | Overloads math operators. |
| **Equality** | `__eq__`, `__lt__` | `==`, `<` | Overloads comparison operators. |
| **Callable** | `__call__(self)` | `obj()` | Allows object to be invoked like a function. |
