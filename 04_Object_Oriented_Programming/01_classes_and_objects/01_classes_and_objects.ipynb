## 1. Object-Oriented Philosophy & Core Definitions

### What is a Class?
> **Definition**: A **Class** is a user-defined blueprint, prototype, or data type that encapsulates data members (state) and member methods (behavior) into a cohesive unit.

### What is an Object?
> **Definition**: An **Object** is a physical, concrete **instance** of a class allocated in computer RAM, possessing its own independent identity (`id()`), state, and behavior.

### Advantages of OOP Architecture:
1. **Security & Data Hiding**: Protect internal object state from unauthorized external mutation.
2. **Code Reusability**: Inherit, specialize, and compose components across modular hierarchies.
3. **Effective Memory Management**: Objects are dynamically allocated on the heap and automatically cleaned up by garbage collection.
4. **Real-World Modeling**: Entities (Students, Bank Accounts, Network Connections) map naturally to classes.

## 2. The 2 Types of Data Members: Instance vs. Class-Level

| Data Member Type | Definition & Placement | Memory Behavior | Access Syntax |
|---|---|---|---|
| **Instance Variables** | Defined inside constructor or methods via `self.variable_name`. | **Unique per object**. Stored in each instance's personal `__dict__`. | `self.var` or `obj.var` |
| **Class-Level Variables** | Defined directly inside the class body, outside all methods. | **Shared across all instances**. Stored once in the class `__dict__`. | `ClassName.var` |

## 3. The 3 Types of Methods: Instance, Class & Static Methods

```
+-------------------------------------------------------------------------+
|                      THE 3 METHOD ARCHITECTURES                         |
|                                                                         |
|  1. Instance Method  --> Takes 'self' (Manipulates individual object)   |
|  2. Class Method     --> Takes 'cls'  (Manipulates shared class state)  |
|  3. Static Method    --> Takes no ref (Pure independent utility helper) |
+-------------------------------------------------------------------------+
```

- **Instance Method**: First parameter must be `self`. Operates on the calling instance's private data.
- **Class Method (`@classmethod`)**: First parameter must be `cls`. Can inspect and modify class-level state or act as alternative constructors.
- **Static Method (`@staticmethod`)**: Bound neither to `self` nor `cls`. Self-contained utility routine.

````python
# The 3 Method Types & Data Members Execution Suite
class StudentManager:
    # Class-Level Data Member (Shared across all students)
    institute_name = "Python Tech Academy"
    total_enrolled = 0

    def __init__(self, sno, sname, course, fee):
        # Instance Data Members (Unique per object)
        self.sno = sno
        self.sname = sname
        self.course = course
        self.fee = fee
        StudentManager.total_enrolled += 1

    # 1. Instance Method: Accesses and displays instance state
    def display_card(self):
        print(f"[{self.institute_name}] ID #{self.sno:<4} | Name: {self.sname:<15} | Course: {self.course:<20} | Fee: Rs. {self.fee:,.2f}")

    # 2. Class Method: Accesses shared class attributes or acts as factory
    @classmethod
    def get_enrollment_summary(cls):
        return f"Total Students Enrolled at {cls.institute_name}: {cls.total_enrolled}"

    @classmethod
    def create_from_csv_string(cls, csv_line):
        """Alternative constructor factory using class method."""
        sno, name, course, fee = csv_line.split(",")
        return cls(int(sno.strip()), name.strip(), course.strip(), float(fee.strip()))

    # 3. Static Method: Independent utility calculation
    @staticmethod
    def calculate_scholarship(fee, percent):
        return fee - (fee * percent / 100.0)

# Instantiating objects
s1 = StudentManager(101, "Aravind Rao", "Full Stack Python", 15000.0)
s2 = StudentManager(102, "Bhavna Patel", "Data Engineering", 18000.0)
s3 = StudentManager.create_from_csv_string("103, Charan Kumar, AI & ML, 22000.0")

s1.display_card()
s2.display_card()
s3.display_card()
print("\nClass Method Summary:", StudentManager.get_enrollment_summary())
print(f"Static Method Utility: Rs. 15,000 with 15% scholarship = Rs. {StudentManager.calculate_scholarship(15000, 15):,.2f}")
````

## 4. Constructor Overloading Simulation

In languages like C++ or Java, multiple constructors with different parameter signatures can exist simultaneously.
In Python, **defining multiple `__init__` methods causes the last definition to overwrite all previous ones**.

We achieve constructor overloading through **Default Parameters** or **Variable Arguments (`*args`, `**kwargs`)**.

````python
# Simulating Constructor Overloading in Python
class Employee:
    def __init__(self, emp_id=None, name=None, salary=30000.0, **extra_details):
        self.emp_id = emp_id if emp_id is not None else 0
        self.name = name if name is not None else "Intern / Guest"
        self.salary = salary
        self.details = extra_details

    def show(self):
        print(f"Emp #{self.emp_id:<4} | Name: {self.name:<15} | Sal: Rs. {self.salary:>8,d} | Extra: {self.details}")

e1 = Employee()                                         # Default constructor
e2 = Employee(201, "Deepa")                             # Partial arguments
e3 = Employee(202, "Eshwar", 85000, city="Hyderabad", dept="DevOps") # Full arguments

e1.show()
e2.show()
e3.show()
````

## 5. Summary & Quick Revision Cheat Sheet

| Component | Decorator | First Param | Operational Scope |
|---|---|---|---|
| **Instance Method** | None | `self` | Accesses & mutates unique instance attributes. |
| **Class Method** | `@classmethod` | `cls` | Accesses shared class variables & factory constructors. |
| **Static Method** | `@staticmethod` | None | Utility calculations isolated from class/instance state. |
| **Instance Variable** | None | `self.x` | Unique per object; stored in `obj.__dict__`. |
| **Class Variable** | None | `cls.x` | Shared across all objects; stored in `Class.__dict__`. |
