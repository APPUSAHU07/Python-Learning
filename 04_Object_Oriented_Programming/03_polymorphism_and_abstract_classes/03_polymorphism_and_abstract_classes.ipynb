## 1. Core Purpose: Polymorphism & Duck Typing

### What is Polymorphism?
> **Etymology**: Greek *Poly* (Many) + *Morph* (Forms).
> **Definition**: The ability of different underlying classes to respond to the **exact same method call** in their own unique, specialized manner.

### Duck Typing in Python:
> *"If it walks like a duck and it quacks like a duck, it's a duck."*
> In Python, a function does not care what explicit class an object belongs to; it only cares whether the object **implements the expected method interface** at runtime.

## 2. Method Overriding & Invoking Base Class Logic

When a Derived Class implements a method already present in its Parent Class, it **overrides** the base implementation.

### The 2 Approaches to Call the Original Overridden Base Logic:
1. **`super().method_name(*args)`**: Standard Pythonic approach respecting MRO.
2. **`ParentClass.method_name(self, *args)`**: Explicit class reference passing `self` manually.

```python
# Method Overriding & Polymorphic Dispatch Suite
class NotificationChannel:
    def send(self, recipient, message):
        print(f"[Base Channel] Storing audit record for {recipient}")

class EmailNotification(NotificationChannel):
    def send(self, recipient, message):
        super().send(recipient, message)  # Approach 1: super()
        print(f"  -> Dispatched EMAIL to '{recipient}': {message}")

class SMSNotification(NotificationChannel):
    def send(self, recipient, message):
        NotificationChannel.send(self, recipient, message)  # Approach 2: Explicit Parent
        print(f"  -> Dispatched SMS to '{recipient}': {message}")

# Polymorphic Dispatch Function (Duck Typing)
def broadcast_alert(channels, user_contact, alert_text):
    print("=" * 55)
    print("             DISPATCHING ALERTS")
    print("=" * 55)
    for channel in channels:
        channel.send(user_contact, alert_text)  # Same method call behaves differently!
    print("=" * 55)

broadcast_alert([EmailNotification(), SMSNotification()], "devops-lead@tech.org", "High CPU Load on Server-01")
```

## 3. Abstract Base Classes (ABC) & Interface Contracts

> **Definition**: An **Abstract Class** is a class that contains one or more **Abstract Methods** (methods declared without an implementation body).
>
> **The Golden Rules of Abstract Classes**:
> 1. Defined by inheriting from `abc.ABC` and decorating methods with `@abstractmethod`.
> 2. **Cannot be instantiated directly** (attempting `obj = AbstractClass()` raises `TypeError`).
> 3. **Mandates Subclass Compliance**: Any subclass MUST implement all abstract methods; otherwise, the subclass remains abstract and cannot be instantiated.

```python
# Abstract Base Class Payment Gateway Suite
from abc import ABC, abstractmethod

class PaymentGateway(ABC):
    def __init__(self, merchant_id):
        self.merchant_id = merchant_id

    @abstractmethod
    def authenticate_merchant(self):
        """Mandatory merchant security handshake."""
        pass

    @abstractmethod
    def process_charge(self, amount, currency):
        """Mandatory charge execution."""
        pass

    # Concrete method shared across all gateways
    def log_audit(self, status, amount):
        print(f"[AUDIT] Merchant {self.merchant_id} | Status: {status} | Amount: Rs. {amount:,.2f}")

# Concrete Implementation 1: UPI Gateway
class UPIGateway(PaymentGateway):
    def authenticate_merchant(self):
        return f"UPI Token verified for {self.merchant_id}"

    def process_charge(self, amount, currency):
        self.log_audit("SUCCESS", amount)
        return f"Processed UPI charge of {currency} {amount:,.2f}"

# Concrete Implementation 2: Credit Card Gateway
class CardGateway(PaymentGateway):
    def authenticate_merchant(self):
        return f"PCI-DSS Token verified for {self.merchant_id}"

    def process_charge(self, amount, currency):
        fee = amount * 0.02  # 2% card fee
        self.log_audit("SUCCESS (FEE APPLIED)", amount + fee)
        return f"Processed Card charge of {currency} {amount + fee:,.2f} (Includes Rs. {fee:.2f} fee)"

# Executing payment processors
upi = UPIGateway("MERCH-UPI-99")
card = CardGateway("MERCH-CARD-44")

print(upi.authenticate_merchant())
print(upi.process_charge(5000.0, "INR"))
print("\n" + card.authenticate_merchant())
print(card.process_charge(10000.0, "INR"))
```

## 4. Summary & Quick Revision Cheat Sheet

| Concept | Implementation Tool | Primary Engineering Role |
|---|---|---|
| **Polymorphism** | Same method name in multiple classes | Uniform dispatch across varying objects. |
| **Duck Typing** | Runtime method check | No explicit inheritance needed if interface matches. |
| **Method Overriding** | Redefining parent method | Specializing base behavior in derived classes. |
| **`super().method()`**| Proxy call to parent | Extends base logic without hardcoding parent class. |
| **Abstract Class** | Inherits `abc.ABC` | Cannot be instantiated; defines mandatory API contract. |
| **`@abstractmethod`** | Decorator from `abc` | Enforces implementation in all concrete subclasses. |
